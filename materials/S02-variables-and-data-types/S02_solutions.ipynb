### Exercise 2: Predict the Type

`10 / 5` is a `float` because `/` always returns a float. `10 // 5` is an `int` because both values are integers.

```python
print(type(2026))     # int
print(type("2026"))   # str
print(type(20.26))    # float
print(type(False))    # bool
print(type(10 / 5))   # float
print(type(10 // 5))  # int
```

## Part B: Intermediate

### Exercise 3: Splitting a Restaurant Bill

```python
bill = 186.40
tip_rate = 0.12
people = 4

tip = bill * tip_rate
total = bill + tip
per_person = total / people

print(f"Tip: {tip:.2f}")
print(f"Total: {total:.2f}")
print(f"Each person pays: {per_person:.2f}")
```

### Exercise 4: Seconds to Hours, Minutes and Seconds

```python
duration_seconds = 7385

hours = duration_seconds // 3600
remaining = duration_seconds % 3600
minutes = remaining // 60
seconds = remaining % 60

print(f"{hours} hours, {minutes} minutes and {seconds} seconds")
```

## Part C: Challenge

### Exercise 5: Savings Growth

```python
start = 5000
rate = 0.04

balance_1 = start * (1 + rate) ** 1
balance_2 = start * (1 + rate) ** 2
balance_3 = start * (1 + rate) ** 3

print(f"Balance after 1 year: {balance_1:.2f}")
print(f"Balance after 2 years: {balance_2:.2f}")
print(f"Balance after 3 years: {balance_3:.2f}")
print(f"Interest earned in 3 years: {balance_3 - start:.2f}")

years = int(input("Number of years: "))
print(f"Balance after {years} years: {start * (1 + rate) ** years:.2f}")
```

## Part D: Cumulative Project: Campus Bookshop

### Exercise 6: The First Receipt

```python
title = "Python for Beginners"
price = 45.0
quantity = 3
tax_rate = 0.15

subtotal = price * quantity
tax = subtotal * tax_rate
total = subtotal + tax

print("Campus Bookshop")
print(f"Book: {title}")
print(f"Quantity: {quantity}")
print("-" * 30)
print(f"Subtotal: {subtotal:.2f}")
print(f"Tax: {tax:.2f}")
print(f"Total: {total:.2f}")
print("-" * 30)
```
