## 1. What Is a Variable?

A **variable** is a name that refers to a value. You create one with the **assignment operator** `=`.

```python
sessions = 52
minutes_per_session = 75

print(sessions * minutes_per_session)
```

Read `sessions = 52` as "store 52 under the name `sessions`". In Python, `=` does not mean "is equal to" as it does in mathematics. It means "take the value on the right and give it the name on the left".

Once a variable exists, any later cell can use it. This is why running cells in order matters.

```python
total_minutes = sessions * minutes_per_session
total_hours = total_minutes / 60

print("Total teaching hours:", total_hours)
```

## 2. Naming Variables

Python has strict **rules** for names. Breaking a rule causes an error.

1. A name can contain letters, digits and underscores.
2. A name cannot start with a digit.
3. A name cannot contain spaces.
4. A name cannot be one of Python's reserved words, such as `if`, `for` or `True`.
5. Names are case sensitive: `price` and `Price` are two different variables.

Python programmers also follow **conventions**, which make code easier to read:

1. Use lowercase words separated by underscores, called snake case: `minutes_per_session`.
2. Choose names that describe the value: `total_price` is better than `tp` or `x`.

| Name | Allowed? | Comment |
| --- | --- | --- |
| `student_count` | Yes | Clear snake case name |
| `studentCount` | Yes | Works, but not the Python convention |
| `2nd_place` | No | Starts with a digit |
| `total price` | No | Contains a space |
| `for` | No | Reserved word |
| `x` | Yes | Allowed, but says nothing about the value |

## 3. Data Types

Every value in Python has a **type**, which decides what you can do with it. Four types cover most of what you need in Stage 1.

| Type | Name | Examples | Used for |
| --- | --- | --- | --- |
| `int` | Integer | `52`, `-3`, `0` | Whole numbers, counts |
| `float` | Floating-point number | `12.5`, `0.15`, `-2.0` | Numbers with a decimal part, measurements |
| `str` | String | `"Python"`, `'Amman'`, `"52"` | Text |
| `bool` | Boolean | `True`, `False` | Yes or no answers |

The function `type()` tells you the type of a value.

```python
city = "Amman"
temperature = 24.5
students = 30
is_open = True

print(type(city))
print(type(temperature))
print(type(students))
print(type(is_open))
```

Notice that `"52"` with quotation marks is a string, not a number. It looks like a number, but Python treats it as text.

```python
print(type(52))
print(type("52"))
```

## 4. Updating Variables

A variable can be given a new value at any time. The new value replaces the old one.

```python
enrolled = 30
print("Enrolled:", enrolled)

enrolled = enrolled + 5
print("Enrolled after new sign-ups:", enrolled)
```

`enrolled = enrolled + 5` looks strange in mathematics, but in Python it means: take the current value of `enrolled`, add 5, and store the result back under the same name.

This pattern is so common that Python has a short form, `+=`. The same short form exists for the other operators: `-=`, `*=` and `/=`.

```python
enrolled += 2
print("Enrolled after two more:", enrolled)
```

## 5. Arithmetic Operators

In Session 01 you used `+`, `-`, `*` and `/`. Python has three more operators that are very useful when working with data.

| Operator | Meaning | Example | Result |
| --- | --- | --- | --- |
| `/` | Division | `135 / 60` | `2.25` |
| `//` | Floor division: how many whole times | `135 // 60` | `2` |
| `%` | Remainder after floor division | `135 % 60` | `15` |
| `**` | Power | `2 ** 3` | `8` |

Floor division and remainder work as a pair. Together they split a quantity into whole units and what is left over.

```python
movie_minutes = 135

hours = movie_minutes // 60
minutes = movie_minutes % 60

print("Hours:", hours)
print("Minutes:", minutes)
```

Two more facts about numbers:

1. Division with `/` always gives a `float`, even when the result is whole: `10 / 2` is `5.0`.
2. When you mix an `int` and a `float` in one calculation, the result is a `float`.

The order of operations is: `**` first, then `*`, `/`, `//` and `%`, then `+` and `-`. Use parentheses whenever the order is not obvious.

```python
print(10 / 2)
print(3 + 2.0)
print(2 + 3 ** 2)
print((2 + 3) ** 2)
```

## 6. Working with Strings

Some operators also work on strings, with a different meaning:

1. `+` joins two strings. This is called **concatenation**.
2. `*` repeats a string a number of times.
3. `len()` counts the characters in a string, including spaces.

You will study strings in depth in Session 08.

```python
first_name = "Lina"
last_name = "Haddad"

full_name = first_name + " " + last_name
print(full_name)
print("-" * 20)
print("Characters in the name:", len(full_name))
```

## 7. Converting Between Types

Python does not guess what you mean when you mix text and numbers. Run the next cell.

```python
age = "29"
print(age + 1)
```

The `TypeError` says that Python cannot add a string and a number. To fix it, **convert** the value to the type you need.

| Function | Converts to | Example | Result |
| --- | --- | --- | --- |
| `int()` | Integer | `int("29")` | `29` |
| `float()` | Float | `float("12.5")` | `12.5` |
| `str()` | String | `str(52)` | `"52"` |

`int()` on a float drops the decimal part without rounding: `int(7.9)` is `7`.

```python
age = "29"
age_next_year = int(age) + 1
print(age_next_year)

print(int(7.9))
print("Session " + str(2))
```

## 8. Clear Output with f-strings

Joining text and numbers with `+` and `str()` quickly becomes hard to read. An **f-string** is a string with the letter `f` before the opening quotation mark. Inside it, any variable or calculation in curly braces `{}` is replaced by its value.

```python
course = "Data Science with Python"
sessions = 52

print(f"{course} has {sessions} sessions.")
print(f"That is {sessions // 3} full weeks at three sessions a week.")
```

f-strings can also control how numbers look. Add `:.2f` after the value to show exactly two decimal places. This is how prices and averages are usually displayed.

```python
price = 12.5
tax_rate = 0.15

print(f"Price with tax: {price * (1 + tax_rate):.2f}")
```

## 9. Reading Input from the User

`input()` shows a message, waits for the user to type something, and returns what they typed. It **always returns a string**, so convert the result before calculating with it.

Run the cell and type your answers in the box that appears.

```python
name = input("What is your name? ")
age = int(input("How old are you? "))

print(f"Hello {name}. In ten years you will be {age + 10}.")
```

## 10. Worked Example: Trip Cost Calculator

A student plans a trip and wants to know the fuel cost and the cost per person. We build the program in three steps.

**Step 1.** Store the inputs in variables with clear names.

```python
distance_km = 420
litres_per_100_km = 6.5
price_per_litre = 2.4
passengers = 3
```

**Step 2.** Calculate the results, one idea per line.

```python
litres_needed = distance_km / 100 * litres_per_100_km
total_cost = litres_needed * price_per_litre
cost_per_person = total_cost / passengers
```

**Step 3.** Display the results with f-strings and two decimal places.

```python
print(f"Fuel needed: {litres_needed:.2f} litres")
print(f"Total fuel cost: {total_cost:.2f}")
print(f"Cost per person: {cost_per_person:.2f}")
```

Compare this with the road trip exercise from Session 01. If the distance changes, you now edit one line instead of three, and every result updates.

## Common Mistakes

1. **Using a variable before creating it.** Python reports a `NameError`. Run the cell that creates it first.
2. **Forgetting that `input()` returns a string.** Convert with `int()` or `float()` before calculating.
3. **Confusing `/` and `//`.** `/` gives the exact result, `//` gives the whole number of times.
4. **Forgetting the `f` before an f-string.** Without it, Python prints the curly braces as plain text.
