### In-Class Exercise 2: Show Data and a Chart

```python
%%writefile explore.py
import pandas as pd
import streamlit as st

DATA_URL = "https://raw.githubusercontent.com/mohammadyahoo/ds-python-course/main/data/"

@st.cache_data
def load_data():
    return pd.read_csv(DATA_URL + "apartments.csv")

flats = load_data()
st.title("Apartment Explorer")
st.dataframe(flats.head())
district = st.selectbox("District", sorted(flats["district"].unique()))
subset = flats[flats["district"] == district]
st.metric("Average price", f"{subset['price_k'].mean():.0f} thousand")
st.bar_chart(subset.groupby("rooms")["price_k"].mean())
```

### Exercise 3: Churn Risk App

`total_charges` is estimated from charges times tenure, since a new customer record would not have it yet. Passing the text `None` for internet works because the imputer turned missing internet values into the category `None` during training.

```python
%%writefile churn_app.py
import joblib
import pandas as pd
import streamlit as st

@st.cache_resource
def load_model():
    return joblib.load("models/churn_model.joblib")

model = load_model()
st.title("Churn Risk")
gender = st.selectbox("Gender", ["Female", "Male"])
senior = st.checkbox("Senior citizen")
tenure = st.slider("Months as a customer", 0, 72, 12)
contract = st.selectbox("Contract", ["Month-to-month", "One year", "Two year"])
internet = st.selectbox("Internet", ["Fiber", "DSL", "None"])
charges = st.slider("Monthly charges", 15.0, 120.0, 70.0)
calls = st.number_input("Support calls", 0, 10, 1)
payment = st.selectbox("Payment method", ["Electronic check", "Card", "Bank transfer", "Cash"])
paperless = st.checkbox("Paperless billing", value=True)

row = pd.DataFrame([{"gender": gender, "senior": senior, "tenure_months": tenure, "contract": contract,
                     "internet": internet, "monthly_charges": charges,
                     "total_charges": charges * tenure, "support_calls": calls,
                     "payment_method": payment, "paperless_billing": paperless}])
risk = float(model.predict_proba(row)[0, 1])
st.metric("Risk of leaving", f"{risk:.0%}")
if risk >= 0.3:
    st.warning("Call this customer")
```

### Exercise 5: Compare with Similar Flats

`@st.cache_data` caches data such as DataFrames; `@st.cache_resource` caches shared objects such as models.

```python
@st.cache_data
def load_flats():
    return pd.read_csv("data/apartments.csv")

flats = load_flats()
same = flats[flats["district"] == district].copy()
same["gap"] = (same["area_m2"] - area).abs()
st.subheader("Similar flats in the training data")
st.dataframe(same.nsmallest(5, "gap")[["area_m2", "rooms", "building_age", "price_k"]])
st.caption("Real examples let users check the estimate against actual prices.")
```
