## Setup

Run this cell first. The lecture builds a short example report on the bike rentals data.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
DATA_URL = "https://raw.githubusercontent.com/mohammadyahoo/ds-python-course/main/data/"
bikes = pd.read_csv(DATA_URL + "bike_rentals.csv")
bikes["date"] = pd.to_datetime(bikes["date"])
```

## 2. The Analysis Workflow

```python
print(bikes.shape)
print(bikes["date"].min().date(), "to", bikes["date"].max().date())
print(bikes.isna().sum().sum(), "missing values")
print(bikes["date"].duplicated().sum(), "repeated dates")
```

## 3. Answering a Question Well

### Question 1: How much did rentals grow from 2023 to 2024?

A good answer has three parts: the **evidence** (a chart or a table), the **number** that answers the question, and one sentence that **interprets** it.

```python
bikes["year"] = bikes["date"].dt.year
bikes["month"] = bikes["date"].dt.month
yearly = bikes.groupby("year")["rentals"].mean()
growth = (yearly[2024] / yearly[2023] - 1) * 100
print(yearly.round(0))
print(f"Growth: {growth:.1f} percent")
```

```python
monthly = bikes.groupby(["month", "year"])["rentals"].mean().unstack()

ax = monthly.plot(marker="o", figsize=(8, 4))
ax.set_title("Average daily rentals per month, 2023 and 2024")
ax.set_xlabel("Month")
ax.set_ylabel("Rentals per day")
ax.set_xticks(range(1, 13))
plt.show()
```

**Answer.** Average daily rentals grew by about 25 percent from 2023 to 2024, and every month of 2024 was busier than the same month of 2023. The growth is steady across the year, so it is not explained by one unusual season.

## 4. Writing the Report

| Section | What goes in it | Length |
| --- | --- | --- |
| Title and summary | Three to five findings in plain words | One short paragraph |
| Data | Source, period, rows, columns, quality checks | A few lines |
| Question 1 to 4 | Question, chart, number, interpretation | One screen each |
| Caveats | What the data cannot tell us, and why | Three bullets |
| Next steps | What you would analyse with more time or data | Two bullets |

## 5. How the Report Is Graded

| Criterion | Points | What earns full points |
| --- | --- | --- |
| Questions | 15 | Four specific, answerable questions |
| Data quality | 15 | Checks run, problems handled and explained |
| Analysis | 25 | Correct pandas, suitable groupings and comparisons |
| Charts | 20 | Right chart types, titles, labels, readable |
| Story | 15 | Clear summary, answers in plain words, caveats |
| Reproducible | 10 | Runs from top to bottom without errors |

## Common Mistakes

| Symptom | Mistake | Fix |
| --- | --- | --- |
| No point | Charts that answer no question | Delete them; every chart must support an answer. |
| Wall of code | Code cells with no text between them | Add a text cell before and after every result. |
| Overclaiming | Writing causes when the data shows patterns | Use words like "linked to" and "in this data". |
| Broken notebook | Cells that only work in a special order | Restart and run all before you submit. |
