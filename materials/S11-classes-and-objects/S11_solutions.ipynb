## In-Class Exercises

### In-Class Exercise 1: A Student Class

```python
class Student:
    def __init__(self, name, scores=None):
        self.name = name
        self.scores = scores if scores is not None else []

    def add_score(self, score):
        self.scores.append(score)

    def average(self):
        return sum(self.scores) / len(self.scores)

s = Student("Lina", [88, 92])
s.add_score(79)
s.add_score(95)
print(s.name, round(s.average(), 1))
```

### In-Class Exercise 2: A Bank Account

```python
class BankAccount:
    def __init__(self, owner):
        self.owner = owner
        self.balance = 0

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            print("Not enough money")
            return
        self.balance -= amount

acc = BankAccount("Omar")
acc.deposit(100)
acc.withdraw(30)
acc.withdraw(500)
print(acc.owner, acc.balance)
```

### In-Class Exercise 3: A Rectangle

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

    def is_square(self):
        return self.width == self.height

for r in [Rectangle(3, 4), Rectangle(5, 5), Rectangle(2, 7.5)]:
    print(r.area(), r.perimeter(), r.is_square())
```

### Exercise 1: A Dog Class

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def speak(self):
        return f"{self.name} says woof"

a = Dog("Rex", 4)
b = Dog("Luna", 7)
print(a.speak())
print(b.speak())
print("Older:", max([a, b], key=lambda d: d.age).name)
```

### Exercise 2: A Temperature Class

```python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    def to_fahrenheit(self):
        return self.celsius * 9 / 5 + 32

    def to_kelvin(self):
        return self.celsius + 273.15

t = Temperature(25)
print(t.celsius, t.to_fahrenheit(), t.to_kelvin())
```

### Exercise 3: A Shopping Cart

```python
class Cart:
    def __init__(self):
        self.items = {}

    def add(self, name, price, quantity=1):
        old_qty = self.items.get(name, (price, 0))[1]
        self.items[name] = (price, old_qty + quantity)

    def remove(self, name):
        self.items.pop(name, None)

    def total(self):
        return sum(p * q for p, q in self.items.values())

    def count(self):
        return sum(q for p, q in self.items.values())

cart = Cart()
cart.add("Notebook", 3.5, 4)
cart.add("Pen", 1.2, 10)
cart.add("Bag", 25.0)
cart.remove("Bag")
print(cart.count(), round(cart.total(), 2))
```

### Exercise 4: A Quiz Question

```python
class Question:
    def __init__(self, text, options, correct):
        self.text = text
        self.options = options
        self.correct = correct

    def show(self):
        print(self.text)
        for i, opt in enumerate(self.options, start=1):
            print(f"  {i}. {opt}")

    def check(self, choice):
        return choice - 1 == self.correct

q1 = Question("Which keyword defines a class?", ["def", "class", "object"], 1)
q2 = Question("What is the first parameter of a method?", ["self", "this", "me"], 0)
q1.show()
print(q1.check(2), q2.check(3))
```

### Exercise 5: A Simple Inventory Tracker

```python
class Product:
    def __init__(self, name, price, stock):
        self.name, self.price, self.stock = name, price, stock

    def sell(self, qty):
        if qty > self.stock:
            return 0
        self.stock -= qty
        return qty * self.price

class Store:
    def __init__(self):
        self.products = []

    def add_product(self, product):
        self.products.append(product)

    def find(self, name):
        for p in self.products:
            if p.name == name:
                return p
        return None

    def sell(self, name, qty):
        p = self.find(name)
        if p is None:
            print("No such product:", name)
            return 0
        revenue = p.sell(qty)
        if revenue == 0:
            print(f"Not enough {name}")
        return revenue

    def report(self):
        for p in self.products:
            print(f"{p.name:<10} {p.stock:>3}")
        print("Stock value:", round(sum(p.price * p.stock for p in self.products), 2))

store = Store()
for p in [Product("Pen", 1.2, 50), Product("Notebook", 3.5, 20), Product("Bag", 25.0, 4)]:
    store.add_product(p)
revenue = store.sell("Pen", 10) + store.sell("Bag", 5) + store.sell("Notebook", 3) + store.sell("Lamp", 1)
print("Revenue:", revenue)
store.report()
```

## Part D: Cumulative Project: Campus Bookshop

### Exercise 6: The Book Class

```python
class Book:
    def __init__(self, title, price, stock, category):
        self.title, self.price, self.stock, self.category = title, price, stock, category

    def sell(self, quantity, member=False):
        if quantity > self.stock:
            print(f"Only {self.stock} copies of {self.title} available")
            return 0
        rate = 0.10 if quantity >= 10 else 0
        if member:
            rate += 0.05
        self.stock -= quantity
        return round(self.price * quantity * (1 - rate) * 1.15, 2)

    def is_low_stock(self, limit=5):
        return self.stock < limit

    def describe(self):
        return f"{self.title} ({self.category}): {self.stock} copies at {self.price:.2f}"

books = [Book("Python Basics", 45.0, 12, "Programming"),
         Book("Data Science Handbook", 62.5, 3, "Data"),
         Book("Statistics Made Simple", 38.0, 0, "Data"),
         Book("SQL in a Week", 29.9, 7, "Programming")]

print("To pay:", books[0].sell(12, member=True))
print("To pay:", books[1].sell(4))
for b in books:
    print(b.describe())
print("Low stock:", [b.title for b in books if b.is_low_stock()])
```
