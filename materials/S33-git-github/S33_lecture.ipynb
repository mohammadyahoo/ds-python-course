## 1. Why Version Control?

| Word | Meaning | Everyday comparison |
| --- | --- | --- |
| Repository (repo) | A project folder that Git tracks | A project binder |
| Commit | A saved snapshot with a message | A dated photo of the binder |
| Staging | Choosing which changes go into the next commit | Putting pages in the tray |
| Branch | A separate line of work | A copy to experiment on |
| Remote | The copy on GitHub | The binder in the cloud |
| Push / pull | Send commits up / bring them down | Upload / download |

## Common Mistakes

| Symptom | Mistake | Fix |
| --- | --- | --- |
| Secrets in Git | Committing passwords or tokens | Never commit secrets; once pushed, assume they are public. |
| Huge files | Committing a 500 MB dataset | Keep data small or link to it; GitHub limits files to 100 MB. |
| Meaningless messages | `git commit -m "update"` | Say what the commit does. |
| No README | A repository with only a notebook | Add a README that explains the project. |

## Summary
