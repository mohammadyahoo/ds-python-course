## 2. Maths on Whole Arrays

Arithmetic between an array and a number applies to **every item**. Arithmetic between two arrays of the same length works **item by item**. This is called **vectorisation**, and it is the main reason NumPy exists: the loop happens inside NumPy, in fast compiled code, instead of in Python.

```python
prices = np.array([45.0, 29.9, 62.5, 38.0])
copies = np.array([3, 10, 1, 4])

print(prices * 1.15)
print(prices * copies)
print((prices * copies).sum())
print(np.round(prices * 1.15, 2))
```

```python
temps_c = np.array([18.5, 22.0, 25.5, 31.0, 16.0])

temps_f = temps_c * 9 / 5 + 32
print(temps_f)
print(np.sqrt(np.array([4, 9, 16])))
print(np.log10(np.array([10, 100, 1000])))
```

## 3. Two-Dimensional Arrays

An array can have rows and columns. A 2D array is like a spreadsheet of numbers: `shape` gives `(rows, columns)`. Use `array[row, column]` to pick one value. A colon on its own means "all", so `array[:, 1]` is the whole second column.

```python
sales = np.array([
    [12, 15, 9],
    [20, 18, 25],
    [7, 11, 14],
    [16, 13, 19],
])

print(sales.shape)
print(sales[1, 2])
print(sales[0])
print(sales[:, 1])
```

```python
print("Total:", sales.sum())
print("Per branch:", sales.sum(axis=0))
print("Per week:", sales.sum(axis=1))
print("Best week per branch:", sales.max(axis=0))
```

## 4. Selecting with Masks

A comparison such as `scores >= 60` is applied to every item and gives an array of `True` and `False`, called a **boolean mask**. Putting the mask inside square brackets keeps only the items where it is `True`. This is exactly how pandas filters rows in Session 17.

```python
scores = np.array([72, 45, 88, 59, 95, 61, 38])

passed = scores >= 60
print(passed)
print(scores[passed])
print(passed.sum(), "students passed")
print(scores[(scores >= 60) & (scores < 90)])
```

```python
scores = np.array([72, 45, 88, 59, 95, 61, 38])

capped = np.where(scores > 90, 90, scores)
labels = np.where(scores >= 60, "pass", "fail")
print(capped)
print(labels)
```

## 5. Statistics and Random Data

```python
scores = np.array([72, 45, 88, 59, 95, 61, 38])

print("Mean:", scores.mean().round(2))
print("Median:", np.median(scores))
print("Std:", scores.std().round(2))
print("Min, max:", scores.min(), scores.max())
print("Position of max:", scores.argmax())
```

```python
rng = np.random.default_rng(42)

dice = rng.integers(1, 7, size=10)
heights = rng.normal(170, 8, size=5).round(1)
print(dice)
print(heights)
print(rng.choice(["A", "B", "C"], size=6))
```

## Common Mistakes

| Symptom | Mistake | Fix |
| --- | --- | --- |
| Unexpected joining | Adding two lists instead of two arrays | Lists join with `+`; convert with `np.array` to add item by item. |
| ValueError | Using `and` or `or` with arrays | Use `&` and `|`, with each condition in brackets. |
| ValueError | Item-by-item maths on arrays of different lengths | Check `shape` first; both arrays must line up. |
| Wrong axis | Using `axis=1` when you wanted one result per column | `axis=0` gives one result per column. |

## Summary

- NumPy arrays hold numbers of one type, with a `shape`.
- Maths on arrays applies to every item: no loops needed.
- 2D arrays use `[row, column]`; `:` means all.
- `axis=0` gives results per column; `axis=1` per row.
