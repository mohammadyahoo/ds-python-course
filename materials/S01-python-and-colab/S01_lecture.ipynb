## 1. What Is Python?

A **program** is a list of instructions that a computer follows, one after another, from top to bottom. A **programming language** is the set of words and rules we use to write those instructions.

Python is a programming language that is popular for three reasons:

1. **It is readable.** Python code looks close to plain English, so it is easier to learn and to review.
2. **It is general purpose.** The same language is used for websites, automation, science and finance.
3. **It has strong data tools.** Libraries such as pandas and scikit-learn, which you will use later in this course, make Python the most common language in data science.

## 3. Notebooks and Cells

A Colab file is called a **notebook**. A notebook is made of **cells**, and there are two kinds:

| Cell type | What it holds | What happens when you run it |
| --- | --- | --- |
| Text cell | Explanations written in Markdown | The text is formatted |
| Code cell | Python instructions | Python runs the code and shows the result below the cell |

To run a cell, click inside it and press **Shift + Enter**. Run the code cell below.

```python
print("Hello, Data Science!")
```

The number that appears in brackets next to a code cell shows the order in which cells were run. Notebooks remember what has already been run, so **run cells from top to bottom**. If something behaves strangely, use **Runtime, then Restart and run all** to start again from a clean state.

### Writing in text cells

| You type | You get |

| `# Title` | A large heading |
| `## Section` | A smaller heading |
| `**important**` | **important** |
| `- item` | A bullet point |

## 4. Displaying Output with print()

`print()` displays whatever you put between its parentheses. Text must be inside quotation marks. Numbers are written without them.

```python
print("Welcome to the course")
print(2026)
print(7 + 5)
```

Text inside quotation marks is called a **string**. Python shows a string exactly as written, while it calculates anything that is not in quotes. Compare the two lines below.

```python
print("7 + 5")
print(7 + 5)
```

You can print several values on one line by separating them with commas. Python puts a space between them.

```python
print("Sessions in the course:", 52)
print("Weeks:", 18, "Levels:", 3)
```

## 5. Python as a Calculator

Python follows the usual order of operations: multiplication and division before addition and subtraction. Parentheses change that order.

| Operation | Symbol | Example | Result |
| --- | --- | --- | --- |
| Addition | `+` | `10 + 4` | 14 |
| Subtraction | `-` | `10 - 4` | 6 |
| Multiplication | `*` | `10 * 4` | 40 |
| Division | `/` | `10 / 4` | 2.5 |

```python
print("Teaching minutes per week:", 3 * 75)
print("Average of 70 and 90:", (70 + 90) / 2)
print("Without parentheses:", 70 + 90 / 2)
```

The last two lines give different results. Without parentheses, Python divides 90 by 2 first and then adds 70.

## 6. Comments

A line that starts with `#` is a **comment**. Python ignores it. Comments are notes for people who read your code, including you in a few weeks.

A good comment explains **why** a line exists, not what it does word for word.

```python
# Each session lasts 75 minutes and there are 52 sessions
print("Total teaching minutes:", 52 * 75)

print("Total teaching hours:", 52 * 75 / 60)  # a comment can also follow code
```

## 7. Reading Error Messages

Every programmer sees errors every day. An error message is Python telling you exactly what it could not understand. Read the **last line** first: it gives the type of error and a short explanation.

Run the cell below. A closing parenthesis is missing.

```python
print("Hello"
```

The error type is `SyntaxError`, which means the code breaks Python's grammar rules. Now run the next cell.

```python
Print("Hello")
```

This time the error is `NameError: name 'Print' is not defined`. Python is **case sensitive**: `print` and `Print` are different names, and only `print` exists.

When you meet an error, follow three steps:

1. Read the last line to learn the error type.
2. Find the line number that Python points to.
3. Compare that line with a version you know works.
