## In-Class Exercises

### In-Class Exercise 1: Shapes

```python
import math

class Shape:
    def __init__(self, name):
        self.name = name

    def area(self):
        return 0

class Circle(Shape):
    def __init__(self, radius):
        super().__init__("circle")
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width, height):
        super().__init__("rectangle")
        self.width, self.height = width, height

    def area(self):
        return self.width * self.height

for s in [Circle(2), Rectangle(3, 4)]:
    print(s.name, round(s.area(), 2))
```

### In-Class Exercise 2: Employees

```python
class Employee:
    def __init__(self, name, salary):
        self.name, self.salary = name, salary

    def annual_pay(self):
        return 12 * self.salary

class Manager(Employee):
    def __init__(self, name, salary, bonus):
        super().__init__(name, salary)
        self.bonus = bonus

    def annual_pay(self):
        return super().annual_pay() + self.bonus

for e in [Employee("Lina", 2000), Manager("Omar", 3000, 5000)]:
    print(e.name, e.annual_pay())
```

### In-Class Exercise 3: Animal Sounds

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def sound(self):
        return "..."

class Dog(Animal):
    def sound(self):
        return "woof"

class Cat(Animal):
    def sound(self):
        return "meow"

class Cow(Animal):
    def sound(self):
        return "moo"

for a in [Dog("Rex"), Cat("Luna"), Cow("Daisy")]:
    print(f"{a.name} says {a.sound()}")
```

### In-Class Exercise 4: Is a, or has a?

```python
# Car has an Engine: composition.
# Student is a Person: inheritance.
# Library has Books: composition.
# SavingsAccount is a BankAccount: inheritance.

class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner, self.balance = owner, balance

class SavingsAccount(BankAccount):
    def __init__(self, owner, balance=0, rate=0.03):
        super().__init__(owner, balance)
        self.rate = rate

    def add_interest(self):
        self.balance = round(self.balance * (1 + self.rate), 2)

acc = SavingsAccount("Sara", 1000)
acc.add_interest()
print(acc.owner, acc.balance)
```

### Exercise 1: Vehicles

```python
class Vehicle:
    def __init__(self, brand, wheels):
        self.brand, self.wheels = brand, wheels

    def describe(self):
        return f"{self.brand} with {self.wheels} wheels"

class Car(Vehicle):
    def __init__(self, brand):
        super().__init__(brand, 4)

class Motorbike(Vehicle):
    def __init__(self, brand):
        super().__init__(brand, 2)

print(Car("Toyota").describe())
print(Motorbike("Honda").describe())
```

### Exercise 2: Check the Family

```python
class Vehicle:
    def __init__(self, brand, wheels):
        self.brand, self.wheels = brand, wheels

class Car(Vehicle):
    def __init__(self, brand):
        super().__init__(brand, 4)

class Motorbike(Vehicle):
    def __init__(self, brand):
        super().__init__(brand, 2)

c = Car("Kia")
print(isinstance(c, Car))        # True: c was made from Car
print(isinstance(c, Vehicle))    # True: a Car is a Vehicle
print(isinstance(c, Motorbike))  # False: a Car is not a Motorbike
```

### Exercise 3: Payment Methods

```python
class Payment:
    def __init__(self, amount):
        self.amount = amount

    def fee(self):
        return 0

    def total(self):
        return round(self.amount + self.fee(), 2)

class CardPayment(Payment):
    def fee(self):
        return self.amount * 0.02

class TransferPayment(Payment):
    def fee(self):
        return 1.5

for p in [Payment(200), CardPayment(200), TransferPayment(200)]:
    print(type(p).__name__, p.total())
```

### Exercise 4: Course Members

```python
class Person:
    def __init__(self, name, email):
        self.name, self.email = name, email

    def contact(self):
        return f"{self.name} <{self.email}>"

class Student(Person):
    def __init__(self, name, email, student_id):
        super().__init__(name, email)
        self.student_id = student_id

    def contact(self):
        return super().contact() + f", student {self.student_id}"

class Instructor(Person):
    def __init__(self, name, email, subject):
        super().__init__(name, email)
        self.subject = subject

    def contact(self):
        return super().contact() + f", teaches {self.subject}"

people = [Student("Lina", "lina@uni.edu", "S101"), Student("Omar", "omar@uni.edu", "S102"),
          Instructor("Mohammad", "mohammad@uni.edu", "Data Science")]
for p in people:
    print(p.contact())
```

### Exercise 5: A Tiny Model Family

Every model has the same `fit` and `predict` interface, so the loop treats them the same way. scikit-learn works exactly like this from Session 31.

```python
class Model:
    def fit(self, xs, ys):
        pass

    def predict(self, x):
        raise NotImplementedError("each model must define predict")

class MeanModel(Model):
    def fit(self, xs, ys):
        self.mean = sum(ys) / len(ys)

    def predict(self, x):
        return self.mean

class LineModel(Model):
    def fit(self, xs, ys):
        self.slope = (ys[-1] - ys[0]) / (xs[-1] - xs[0])
        self.intercept = ys[0] - self.slope * xs[0]

    def predict(self, x):
        return self.intercept + self.slope * x

xs, ys = [1, 2, 3, 4, 5], [3, 5, 7, 9, 11]
for m in [MeanModel(), LineModel()]:
    m.fit(xs, ys)
    print(type(m).__name__, m.predict(6))
```

## Part D: Cumulative Project: Campus Bookshop

### Exercise 6: Three Kinds of Products

```python
class Product:
    def __init__(self, title, price):
        self.title, self.price = title, price

    def shipping(self):
        return 5.0

    def total(self, quantity):
        return round(self.price * quantity + self.shipping(), 2)

    def __str__(self):
        return f"{self.title} ({self.price:.2f})"

class PrintedBook(Product):
    def __init__(self, title, price, stock):
        super().__init__(title, price)
        self.stock = stock

    def sell(self, quantity):
        if quantity > self.stock:
            print(f"Only {self.stock} copies of {self.title}")
            return 0
        self.stock -= quantity
        return self.total(quantity)

class EBook(Product):
    def __init__(self, title, price, size_mb):
        super().__init__(title, price)
        self.size_mb = size_mb

    def shipping(self):
        return 0.0

class AudioBook(Product):
    def __init__(self, title, price, minutes):
        super().__init__(title, price)
        self.minutes = minutes

    def shipping(self):
        return 0.0

    def __str__(self):
        return super().__str__() + f", {self.minutes // 60}h {self.minutes % 60}min"

class Shop:
    def __init__(self, name):
        self.name, self.products = name, []

    def add(self, product):
        self.products.append(product)

    def catalogue(self):
        print(self.name)
        for p in sorted(self.products, key=lambda p: p.price):
            print("  ", p)

    def order_total(self, items):
        return round(sum(p.total(q) for p, q in items), 2)

shop = Shop("Campus Bookshop")
items = [PrintedBook("Python Basics", 45.0, 12), PrintedBook("SQL in a Week", 29.9, 2),
         EBook("Python Basics (PDF)", 19.0, 12.5), EBook("Statistics (PDF)", 15.0, 8.0),
         AudioBook("Python Basics (Audio)", 25.0, 310), AudioBook("Data Stories (Audio)", 18.0, 95)]
for p in items:
    shop.add(p)
shop.catalogue()
print("Order total:", shop.order_total([(items[0], 2), (items[2], 1), (items[4], 1)]))
print(items[1].sell(3))
```
