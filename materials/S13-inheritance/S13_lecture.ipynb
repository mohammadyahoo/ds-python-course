## Where We Are

The bookshop now sells three kinds of products: printed books, e-books and audiobooks. They all have a title and a price, and they can all be sold. But printed books need shipping and stock, e-books have a file size, and audiobooks have a length in minutes.

Writing three separate classes would copy the same code three times. **Inheritance** lets a new class start from an existing one and add or change only what is different. scikit-learn is built this way: every model, from linear regression to random forests, inherits the same `fit` and `predict` design.

```python
class PrintedBook:
    def __init__(self, title, price):
        self.title, self.price = title, price

class EBook:
    def __init__(self, title, price):
        self.title, self.price = title, price
```

## 1. Parent and Child Classes

To inherit, put the parent class in brackets after the child's name: `class EBook(Product):`. The child **inherits** every attribute and method of the parent. The parent is also called the base class or superclass; the child is the subclass.

```python
class Product:
    def __init__(self, title, price):
        self.title = title
        self.price = price

    def describe(self):
        return f"{self.title} costs {self.price:.2f}"

class EBook(Product):
    pass

e = EBook("Python Basics (PDF)", 19.0)
print(e.describe())
print(isinstance(e, EBook), isinstance(e, Product))
```

## 2. Extending with super()

A child class often needs extra data. It defines its own `__init__`, and calls the parent's `__init__` with `super().__init__(...)` to set up the shared attributes, instead of repeating that code.

```python
class EBook(Product):
    def __init__(self, title, price, size_mb):
        super().__init__(title, price)
        self.size_mb = size_mb

class PrintedBook(Product):
    def __init__(self, title, price, stock, weight_kg):
        super().__init__(title, price)
        self.stock = stock
        self.weight_kg = weight_kg

e = EBook("Python Basics (PDF)", 19.0, 12.5)
p = PrintedBook("Python Basics", 45.0, 12, 0.6)
print(e.title, e.size_mb)
print(p.title, p.stock, p.weight_kg)
```

## 3. Overriding Methods

If a child defines a method with the same name as one in the parent, the child's version is used for its objects. This is called **overriding**. Inside the new version, `super().method()` can still call the parent's version, so you can extend it instead of rewriting it.

```python
class Product:
    def __init__(self, title, price):
        self.title, self.price = title, price

    def shipping(self):
        return 5.0

    def describe(self):
        return f"{self.title}: {self.price:.2f} + {self.shipping():.2f} shipping"

class EBook(Product):
    def shipping(self):
        return 0.0

class AudioBook(Product):
    def __init__(self, title, price, minutes):
        super().__init__(title, price)
        self.minutes = minutes

    def shipping(self):
        return 0.0

    def describe(self):
        return super().describe() + f", {self.minutes} min"

for item in [Product("Book", 45.0), EBook("PDF", 19.0),
             AudioBook("Audio", 25.0, 310)]:
    print(item.describe())
```

## 4. Polymorphism

Because every product has `shipping` and `describe`, code that works with products does not need to know which kind each one is. It calls the same method, and each object answers in its own way. This is **polymorphism**, from the Greek for "many forms". It is why you can swap one scikit-learn model for another and keep the rest of your code the same.

```python
cart = [Product("Python Basics", 45.0),
        EBook("SQL in a Week (PDF)", 15.0),
        AudioBook("Statistics (Audio)", 22.0, 280)]

total = 0
for item in cart:
    total += item.price + item.shipping()
    print(type(item).__name__, "|", item.describe())

print(f"Total: {total:.2f}")
```

```python
for item in cart:
    if isinstance(item, AudioBook):
        print(item.title, "is", item.minutes, "minutes long")
    elif isinstance(item, EBook):
        print(item.title, "is a download")
    else:
        print(item.title, "is shipped")
```

## 5. Inheritance or Composition?

```python
class Shop:
    def __init__(self, name):
        self.name = name
        self.products = []

    def add(self, product):
        self.products.append(product)

    def catalogue(self):
        for p in sorted(self.products, key=lambda p: p.price):
            print(f"{p.price:>6.2f}  {p.title}")

shop = Shop("Campus Bookshop")
for item in cart:
    shop.add(item)
shop.catalogue()
```

## Common Mistakes

| Symptom | Mistake | Fix |
| --- | --- | --- |
| AttributeError | Forgetting `super().__init__` in the child | Call `super().__init__(...)` first in the child's `__init__`. |
| TypeError | Passing the wrong arguments to `super().__init__` | Pass exactly the parameters the parent's `__init__` expects. |
| Wrong method used | Misspelling the name of an overriding method | The child's method must have the same name as the parent's. |
| Messy design | Inheriting when the child is not really a kind of parent | Use composition for has-a relationships. |

## Summary

- `class Child(Parent):` inherits all of the parent's code.
- `super().__init__(...)` reuses the parent's setup.
- Overriding replaces a method for one kind of object.
- Polymorphism: one method name, many behaviours.
- `isinstance(obj, Class)` checks an object's kind.
- Is a means inheritance; has a means composition.
