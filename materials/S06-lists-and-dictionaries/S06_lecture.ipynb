## 1. Creating Lists and Reading Items

A list is written with square brackets, with items separated by commas. Each item has a position called its **index**. Indexes start at 0, so the first item is at index 0. Negative indexes count from the end: `-1` is the last item.

````python
cities = ["Damascus", "Aleppo", "Homs", "Latakia"]

print(cities[0])
print(cities[2])
print(cities[-1])
print(len(cities))
````

A list can hold any type of value, and the items do not all need the same type. In data work, though, a list usually holds one kind of thing: all numbers, or all names.

## 2. Slicing

A **slice** takes several items at once. `items[start:stop]` gives the items from `start` up to, but not including, `stop`, exactly like `range()`. Leaving out `start` means from the beginning; leaving out `stop` means to the end.

````python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

print(months[0:3])
print(months[3:])
print(months[-2:])
print(months[::2])
````

| Slice | Result | Meaning |
| --- | --- | --- |
| `months[0:3]` | Jan, Feb, Mar | First three items |
| `months[:3]` | Jan, Feb, Mar | Same: start defaults to 0 |
| `months[3:]` | Apr, May, Jun | From index 3 to the end |
| `months[-2:]` | May, Jun | The last two items |
| `months[::-1]` | Jun, May, ..., Jan | The whole list reversed |

## 3. Changing Lists

Lists are **mutable**, which means they can change after they are created. **Methods** are functions that belong to a value; you call them with a dot after the variable name, such as `scores.append(91)`.

````python
cart = ["notebook", "pen"]

cart.append("backpack")
cart.insert(0, "laptop")
cart[2] = "blue pen"
cart.remove("notebook")

print(cart)
print(cart.pop())
print(cart)
````

| Method | What it does | Example |
| --- | --- | --- |
| `append(x)` | Add `x` to the end | `cart.append("pen")` |
| `insert(i, x)` | Add `x` at index `i` | `cart.insert(0, "pen")` |
| `remove(x)` | Delete the first item equal to `x` | `cart.remove("pen")` |
| `pop()` | Remove and return the last item | `last = cart.pop()` |
| `sort()` | Sort the list in place | `scores.sort()` |
| `count(x)` | How many times `x` appears | `votes.count("yes")` |
| `index(x)` | Position of the first `x` | `cities.index("Homs")` |

## 4. Lists and Loops

A `for` loop can go straight through the items of a list. This is the pattern you will use most in data work: go through every value, and calculate something.

````python
prices = [12.5, 8.0, 23.9, 4.5]

total = 0
for price in prices:
    total += price

print("Total:", total)
print("Average:", total / len(prices))
````

````python
temperatures_c = [18.5, 22.0, 25.5, 31.0, 16.0]

temperatures_f = []
for t in temperatures_c:
    temperatures_f.append(round(t * 9 / 5 + 32, 1))

hot_days = [t for t in temperatures_c if t > 24]

print(temperatures_f)
print(hot_days)
````

````python
names = ["Lina", "Omar", "Sara"]
scores = [88, 92, 79]

for i, name in enumerate(names, start=1):
    print(i, name)

for name, score in zip(names, scores):
    print(f"{name}: {score}")
````

## 5. Built-in Functions for Lists

````python
scores = [72, 88, 95, 64, 88, 79]

print("Count:", len(scores))
print("Total:", sum(scores))
print("Average:", round(sum(scores) / len(scores), 1))
print("Highest:", max(scores), "Lowest:", min(scores))
print("Sorted:", sorted(scores))
print("Top three:", sorted(scores, reverse=True)[:3])
````

## 6. Tuples

A **tuple** is written with round brackets. It works like a list for reading, indexing, slicing and looping, but it is **immutable**: once created, it cannot be changed. Use a tuple for a fixed group of values that belong together, such as a point `(x, y)` or a date `(2025, 3, 14)`.

Tuples can be **unpacked** into separate variables in one line. A function can also return several values at once; Python packs them into a tuple.

````python
location = (33.51, 36.29)
lat, lon = location
print("Latitude:", lat, "Longitude:", lon)

def min_max(values):
    return min(values), max(values)

low, high = min_max([72, 88, 95, 64])
print(low, high)
````

Lists find values by **position**. To find the price of one book you would need its index in a list of titles, then the same index in a list of prices. A dictionary stores the pair directly, so you look up the price by the title.

## 7. Creating and Reading Dictionaries

A dictionary is written with curly brackets. Each entry is `key: value`, and entries are separated by commas. Keys are usually strings and must be unique; values can be anything.

````python
prices = {"Espresso": 2.5, "Latte": 3.5, "Muffin": 3.0}

print(prices["Latte"])
print(len(prices))
print("Muffin" in prices)
print("Tea" in prices)
````

````python
prices = {"Espresso": 2.5, "Latte": 3.5, "Muffin": 3.0}

print(prices.get("Tea"))
print(prices.get("Tea", 0))
print(prices.get("Latte", 0))
````

## 8. Changing Dictionaries

````python
stock = {"pen": 40, "notebook": 25}

stock["ruler"] = 15
stock["pen"] = 35
stock["notebook"] -= 5
del stock["ruler"]

print(stock)
print(stock.pop("pen"))
print(stock)
````

## 9. Looping over Dictionaries

````python
scores = {"Lina": 88, "Omar": 92, "Sara": 79}

for name in scores:
    print(name)

print(list(scores.values()))

for name, score in scores.items():
    print(f"{name} scored {score}")
````

## 10. Counting with Dictionaries

Counting how often each value appears is one of the most useful things you can do with data: how many orders per product, how many students per city. A dictionary is perfect for this: the thing you count is the key, and the count is the value.

````python
orders = ["Latte", "Espresso", "Latte", "Muffin", "Latte", "Espresso"]

counts = {}
for item in orders:
    counts[item] = counts.get(item, 0) + 1

print(counts)
````

Python also has a ready-made counter in the `collections` module. You will learn about modules in Session 08, but here is a preview:

```python
from collections import Counter
print(Counter(orders).most_common(2))
```

In Stage 2, pandas does the same with `value_counts()`.

## 11. Records: Dictionaries inside Lists

A dictionary can describe one **record**, such as one student or one product. A list of such dictionaries is a small table: each dictionary is a row and each key is a column. This is exactly the shape of data that pandas turns into a DataFrame in Session 11.

````python
students = [
    {"name": "Lina", "city": "Damascus", "gpa": 3.6},
    {"name": "Omar", "city": "Aleppo", "gpa": 3.1},
    {"name": "Sara", "city": "Damascus", "gpa": 3.9},
]

for s in students:
    print(f"{s['name']:<6} {s['city']:<10} {s['gpa']}")

in_damascus = [s["name"] for s in students if s["city"] == "Damascus"]
print(in_damascus)
````

````python
profile = {
    "name": "Lina",
    "skills": ["Python", "SQL"],
    "address": {"city": "Damascus", "country": "Syria"},
}

print(profile["skills"][0])
print(profile["address"]["city"])
````

## 12. Sets

A **set** is written with curly brackets like a dictionary, but holds single values instead of pairs. A set never contains duplicates, and its items have no order or index. Sets are ideal for removing duplicates and for comparing groups.

````python
cities = ["Damascus", "Aleppo", "Damascus", "Homs", "Aleppo"]
unique_cities = set(cities)
print(unique_cities)
print(len(unique_cities))

python_group = {"Lina", "Omar", "Sara", "Hadi"}
sql_group = {"Omar", "Hadi", "Maya"}
print(python_group & sql_group)
print(python_group | sql_group)
print(python_group - sql_group)
````

| Collection | Written as | Ordered | Changeable | Use it for |
| --- | --- | --- | --- | --- |
| List | `[1, 2, 3]` | Yes | Yes | A sequence of values, in order |
| Tuple | `(1, 2, 3)` | Yes | No | A fixed group of related values |
| Dictionary | `{"a": 1}` | Yes | Yes | Looking values up by a key |
| Set | `{1, 2, 3}` | No | Yes | Unique values and group comparisons |

## Common Mistakes

| Symptom | Mistake | Fix |
| --- | --- | --- |
| IndexError | Using index `len(items)` for the last item | The last item is `items[-1]`, or `items[len(items) - 1]`. |
| Prints None | Storing the result of `scores.sort()` | `sort()` changes the list and returns `None`; use `sorted()` for a copy. |
| KeyError | Reading a key that does not exist | Check with `in`, or use `get(key, default)`. |
| Empty dict | Writing `{}` when you wanted an empty set | `{}` is an empty dictionary; use `set()` for an empty set. |
