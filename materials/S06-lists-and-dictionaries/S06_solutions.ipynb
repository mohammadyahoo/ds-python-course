### In-Class Exercise 4: Letter Counter

```python
word = "statistics"
counts = {}
for ch in word:
    counts[ch] = counts.get(ch, 0) + 1
print(counts)
best = max(counts, key=counts.get)
print(best, counts[best])
```

### Exercise 3: Remove Duplicates

```python
def unique(values):
    result = []
    for v in values:
        if v not in result:
            result.append(v)
    return result

print(unique([3, 1, 3, 2, 1, 5, 2]))
```

### Exercise 4: Word Frequencies

`lambda pair: pair[1]` is a tiny function that returns the second item of each pair, the count. You will see `lambda` again with pandas.

```python
sentence = "the data shows that the model learns from the data"
counts = {}
for w in sentence.lower().split():
    counts[w] = counts.get(w, 0) + 1
for word, n in sorted(counts.items(), key=lambda pair: pair[1], reverse=True):
    print(word, n)
```

### Exercise 5: Group by Category

```python
sales = [("Latte", "Coffee", 3.5), ("Muffin", "Pastry", 3.0), ("Espresso", "Coffee", 2.5),
         ("Croissant", "Pastry", 2.8), ("Latte", "Coffee", 3.5), ("Green Tea", "Tea", 2.5)]
totals = {}
items = {}
for item, category, amount in sales:
    totals[category] = totals.get(category, 0) + amount
    if category not in items:
        items[category] = set()
    items[category].add(item)
print(totals)
print({c: sorted(i) for c, i in items.items()})
```

## Part D: Cumulative Project: Campus Bookshop

### Exercise 6: Bookshop Inventory

```python
titles = ["Python Basics", "Data Science Handbook", "Statistics Made Simple", "SQL in a Week"]
prices = [45.0, 62.5, 38.0, 29.9]
stock = [12, 3, 0, 7]

for t, p, s in zip(titles, prices, stock):
    print(f"{t:<25} {p:>6.2f} {s:>3}")

value = 0
for p, s in zip(prices, stock):
    value += p * s
print(f"Stock value: {value:.2f}")

out_of_stock = [t for t, s in zip(titles, stock) if s == 0]
low_stock = [t for t, s in zip(titles, stock) if s < 5]
print("Out of stock:", out_of_stock)
print("Low stock:", low_stock)

for price, title in sorted(zip(prices, titles), reverse=True):
    print(price, title)
```

### Exercise 7: Bookshop Catalogue

```python
catalogue = {
    "Python Basics": {"price": 45.0, "stock": 12, "category": "Programming"},
    "Data Science Handbook": {"price": 62.5, "stock": 3, "category": "Data"},
    "Statistics Made Simple": {"price": 38.0, "stock": 0, "category": "Data"},
    "SQL in a Week": {"price": 29.9, "stock": 7, "category": "Programming"},
}

def price_of(title):
    book = catalogue.get(title)
    return book["price"] if book else None

def sell(title, quantity):
    book = catalogue[title]
    if book["stock"] < quantity:
        print(f"Sorry, only {book['stock']} copies of {title} available")
        return 0
    book["stock"] -= quantity
    return book["price"] * quantity

print(price_of("SQL in a Week"), price_of("Unknown Book"))
print("To pay:", sell("Python Basics", 2))
print("To pay:", sell("Statistics Made Simple", 1))

for title, book in catalogue.items():
    print(f"{title:<25} {book['stock']}")

per_category = {}
for book in catalogue.values():
    per_category[book["category"]] = per_category.get(book["category"], 0) + book["stock"]
print(per_category)
```
