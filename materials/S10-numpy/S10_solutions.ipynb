### In-Class Exercise 3: Weekly Sales Table

```python
sales = np.array([[12, 15, 9], [20, 18, 25], [7, 11, 14], [16, 13, 19]])
print(sales.mean(axis=0))
print("Best week:", sales.sum(axis=1).argmax() + 1)
print(sales[sales > 15])
```

### Exercise 3: Grades Table

```python
grades = np.array([[78, 85, 92], [65, 70, 58], [90, 94, 88], [55, 62, 71]])
print(grades.mean(axis=1).round(1))
print(grades.mean(axis=0).round(1))
print("Best student:", grades.mean(axis=1).argmax() + 1)
print("Grades below 60:", (grades < 60).sum())
```

## Part C: Challenge

### Exercise 5: Random Walk

```python
rng = np.random.default_rng(3)
steps = rng.choice([-1, 1], size=1000)
position = np.cumsum(steps)
print("Final:", position[-1])
print("Highest:", position.max(), "Lowest:", position.min())
print("Steps above zero:", (position > 0).sum())
```

## Part D: Cumulative Project: Cafe Sales Analysis

### Exercise 6: First Cafe Calculations

`sold * prices` works because NumPy stretches the price row across every day row. This is called broadcasting.

```python
items = np.array(["Espresso", "Latte", "Croissant", "Muffin", "Orange Juice"])
prices = np.array([2.5, 3.5, 2.8, 3.0, 3.8])
sold = np.array([[30, 42, 18, 15, 9],
                 [28, 50, 22, 12, 14],
                 [35, 47, 20, 18, 11],
                 [40, 55, 25, 20, 16]])
revenue = sold * prices
print(revenue)
print("Per day:", revenue.sum(axis=1))
per_item = revenue.sum(axis=0)
print("Per item:", per_item)
print("Best item:", items[per_item.argmax()])
print("Above average:", items[per_item > per_item.mean()])
```
