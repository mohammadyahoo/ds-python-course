## 1. Importing Modules

`import math` loads the whole module; you then write `math.` before each name. `from math import sqrt` brings in one name so you can use it directly. `import statistics as st` gives the module a shorter **alias**. Data scientists use standard aliases, such as `import numpy as np` and `import pandas as pd`, which you will meet in Stage 2.

Put all imports at the top of the notebook, in their own cell.

```python
import math
from math import sqrt, pi
import statistics as st

print(math.floor(7.8), math.ceil(7.2))
print(sqrt(144), round(pi, 4))
print(st.mean([4, 8, 6]))
```

## 3. Randomness and Seeds

The `random` module produces random numbers, picks random items and shuffles lists. Data scientists use randomness to simulate experiments, take samples and split data into training and test sets.

Random numbers on a computer come from a formula that starts at a **seed**. If you set the same seed, you get the same "random" results every time. That makes your work **reproducible**: someone else can run your notebook and get exactly your numbers. You will see this idea again as `random_state=42` in scikit-learn.

```python
import random

random.seed(42)
print(random.randint(1, 6))
print(random.choice(["heads", "tails"]))
print(random.sample(range(1, 50), 6))
print(round(random.random(), 3))

cards = ["A", "K", "Q", "J"]
random.shuffle(cards)
print(cards)
```

## 4. Working with Dates

Dates are everywhere in data: order dates, sign-up dates, sensor times. The `datetime` module knows how many days each month has, handles leap years and can do arithmetic with dates. `strptime` turns text into a date; `strftime` turns a date into text in any format.

```python
from datetime import date, datetime, timedelta

start = date(2025, 9, 14)
end = date(2026, 1, 20)
print((end - start).days, "days")
print(start + timedelta(weeks=18))
print(start.strftime("%A %d %B %Y"))

d = datetime.strptime("14/03/2025 09:30", "%d/%m/%Y %H:%M")
print(d.year, d.month, d.hour, d.strftime("%a"))
```

## 6. Defining a Class

A class is defined with the keyword `class` and a name written in **CapitalisedWords**. Inside, the special method `__init__` runs automatically every time a new object is created. Its job is to store the starting data in **attributes**.

The first parameter of every method is `self`, which means "this particular object". Writing `self.title = title` stores the title inside this object.

```python
class Book:
    def __init__(self, title, price, stock):
        self.title = title
        self.price = price
        self.stock = stock

book = Book("Python Basics", 45.0, 12)
print(book.title)
print(book.price, book.stock)
```

```python
b1 = Book("Python Basics", 45.0, 12)
b2 = Book("SQL in a Week", 29.9, 7)

b2.price = 24.9
print(b1.title, b1.price)
print(b2.title, b2.price)
print(type(b1))
```

## 7. Methods

A **method** is a function defined inside the class. It always takes `self` first, so it can read and change the object's attributes. You call it with a dot: `book.value()`. Python passes the object as `self` for you, so you do not write it in the call.

```python
class Book:
    def __init__(self, title, price, stock):
        self.title = title
        self.price = price
        self.stock = stock

    def stock_value(self):
        return self.price * self.stock

    def describe(self):
        return f"{self.title}: {self.stock} at {self.price:.2f}"

book = Book("Python Basics", 45.0, 12)
print(book.describe())
print(book.stock_value())
```

## 8. Changing State with Methods

The data inside an object at a given moment is its **state**. Instead of changing attributes directly from outside, it is safer to change them through methods. A method can check the rules first, so the object can never end up in an impossible state, such as negative stock.

```python
class Book:
    def __init__(self, title, price, stock=0):
        self.title = title
        self.price = price
        self.stock = stock

    def restock(self, copies):
        self.stock += copies

    def sell(self, copies):
        if copies > self.stock:
            print(f"Only {self.stock} copies left")
            return 0
        self.stock -= copies
        return copies * self.price

book = Book("SQL in a Week", 29.9, 3)
print(book.sell(2), book.stock)
print(book.sell(5), book.stock)
```

## 9. Many Objects Working Together

```python
shelf = [
    Book("Python Basics", 45.0, 12),
    Book("Data Science Handbook", 62.5, 3),
    Book("SQL in a Week", 29.9, 0),
]

total = sum(b.price * b.stock for b in shelf)
print(f"Stock value: {total:.2f}")

print([b.title for b in shelf if b.stock == 0])
most_expensive = max(shelf, key=lambda b: b.price)
print(most_expensive.title)
```

## 10. Printing Objects

When you call `print(obj)` or `str(obj)`, Python looks for a method called `__str__` and prints what it returns. `__repr__` is used when an object is shown inside a list or as the last line of a cell; it should look like the code that would create the object. If you define only one, define `__repr__`.

```python
class Book:
    def __init__(self, title, price, stock=0):
        self.title = title
        self.price = price
        self.stock = stock

    def __str__(self):
        return f"{self.title}, {self.price:.2f} ({self.stock} in stock)"

    def __repr__(self):
        return f"Book({self.title!r}, {self.price}, {self.stock})"

b = Book("Python Basics", 45.0, 12)
print(b)
print([b, Book("SQL in a Week", 29.9)])
```

## Going Further (Optional): Inheritance

This part is optional reading. A class can **inherit** from another class: it gets all the parent's data and methods, and can change some of them. In Project 1 you can earn bonus points by using inheritance, for example an `EBook` that is a kind of `Book` with no shipping cost.

```python
class Product:
    def __init__(self, title, price):
        self.title = title
        self.price = price

    def describe(self):
        return f"{self.title} costs {self.price:.2f}"

class EBook(Product):
    pass

e = EBook("Python Basics (PDF)", 19.0)
print(e.describe())
print(isinstance(e, EBook), isinstance(e, Product))
```

```python
class Product:
    def __init__(self, title, price):
        self.title, self.price = title, price

    def shipping(self):
        return 5.0

    def describe(self):
        return f"{self.title}: {self.price:.2f} + {self.shipping():.2f} shipping"

class EBook(Product):
    def shipping(self):
        return 0.0

class AudioBook(Product):
    def __init__(self, title, price, minutes):
        super().__init__(title, price)
        self.minutes = minutes

    def shipping(self):
        return 0.0

    def describe(self):
        return super().describe() + f", {self.minutes} min"

for item in [Product("Book", 45.0), EBook("PDF", 19.0),
             AudioBook("Audio", 25.0, 310)]:
    print(item.describe())
```

## Common Mistakes

| Symptom | Mistake | Fix |
| --- | --- | --- |
| ModuleNotFoundError | A misspelt module, or one that is not installed | Check the spelling; install it with `!pip install`. |
| Changing results | Random numbers without a seed | Call `random.seed(42)` at the start for repeatable results. |
| TypeError | Forgetting `self` as the first method parameter | Every method starts with `self`: `def sell(self, copies):`. |
| Prints an address | Forgetting `__str__` or `__repr__` | Define `__repr__` in every class you write. |
