## 1. Creating Lists and Reading Items

A list is written with square brackets, with items separated by commas. Each item has a position called its **index**. Indexes start at 0, so the first item is at index 0. Negative indexes count from the end: `-1` is the last item.

```python
cities = ["Damascus", "Aleppo", "Homs", "Latakia"]

print(cities[0])
print(cities[2])
print(cities[-1])
print(len(cities))
```

```python
cities = ["Damascus", "Aleppo", "Homs", "Latakia"]
print(cities[4])
```

A list can hold any type of value, and the items do not all need the same type. In data work, though, a list usually holds one kind of thing: all numbers, or all names.

## 2. Slicing

A **slice** takes several items at once. `items[start:stop]` gives the items from `start` up to, but not including, `stop`, exactly like `range()`. Leaving out `start` means from the beginning; leaving out `stop` means to the end.

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

print(months[0:3])
print(months[3:])
print(months[-2:])
print(months[::2])
```

| Slice | Result | Meaning |
| --- | --- | --- |
| `months[0:3]` | Jan, Feb, Mar | First three items |
| `months[:3]` | Jan, Feb, Mar | Same: start defaults to 0 |
| `months[3:]` | Apr, May, Jun | From index 3 to the end |
| `months[-2:]` | May, Jun | The last two items |
| `months[::-1]` | Jun, May, ..., Jan | The whole list reversed |

## 3. Changing Lists

Lists are **mutable**, which means they can change after they are created. **Methods** are functions that belong to a value; you call them with a dot after the variable name, such as `scores.append(91)`.

```python
cart = ["notebook", "pen"]

cart.append("backpack")
cart.insert(0, "laptop")
cart[2] = "blue pen"
cart.remove("notebook")

print(cart)
print(cart.pop())
print(cart)
```

| Method | What it does | Example |
| --- | --- | --- |
| `append(x)` | Add `x` to the end | `cart.append("pen")` |
| `insert(i, x)` | Add `x` at index `i` | `cart.insert(0, "pen")` |
| `remove(x)` | Delete the first item equal to `x` | `cart.remove("pen")` |
| `pop()` | Remove and return the last item | `last = cart.pop()` |
| `sort()` | Sort the list in place | `scores.sort()` |
| `count(x)` | How many times `x` appears | `votes.count("yes")` |
| `index(x)` | Position of the first `x` | `cities.index("Homs")` |

## 4. Lists and Loops

A `for` loop can go straight through the items of a list. This is the pattern you will use most in data work: go through every value, and calculate something.

```python
prices = [12.5, 8.0, 23.9, 4.5]

total = 0
for price in prices:
    total += price

print("Total:", total)
print("Average:", total / len(prices))
```

```python
temperatures_c = [18.5, 22.0, 25.5, 31.0, 16.0]

temperatures_f = []
for t in temperatures_c:
    temperatures_f.append(round(t * 9 / 5 + 32, 1))

hot_days = [t for t in temperatures_c if t > 24]

print(temperatures_f)
print(hot_days)
```

```python
names = ["Lina", "Omar", "Sara"]
scores = [88, 92, 79]

for i, name in enumerate(names, start=1):
    print(i, name)

for name, score in zip(names, scores):
    print(f"{name}: {score}")
```

## 5. Built-in Functions for Lists

```python
scores = [72, 88, 95, 64, 88, 79]

print("Count:", len(scores))
print("Total:", sum(scores))
print("Average:", round(sum(scores) / len(scores), 1))
print("Highest:", max(scores), "Lowest:", min(scores))
print("Sorted:", sorted(scores))
print("Top three:", sorted(scores, reverse=True)[:3])
```

```python
words = ["data", "python", "ai", "statistics"]

print(sorted(words))
print(sorted(words, key=len))
print("python" in words)
print("java" in words)
```

## 6. Tuples

A **tuple** is written with round brackets. It works like a list for reading, indexing, slicing and looping, but it is **immutable**: once created, it cannot be changed. Use a tuple for a fixed group of values that belong together, such as a point `(x, y)` or a date `(2025, 3, 14)`.

Tuples can be **unpacked** into separate variables in one line. A function can also return several values at once; Python packs them into a tuple.

```python
location = (33.51, 36.29)
lat, lon = location
print("Latitude:", lat, "Longitude:", lon)

def min_max(values):
    return min(values), max(values)

low, high = min_max([72, 88, 95, 64])
print(low, high)
```

## Common Mistakes

| Symptom | Mistake | Fix |
| --- | --- | --- |
| IndexError | Using index `len(items)` for the last item | The last item is `items[-1]`, or `items[len(items) - 1]`. |
| Prints None | Storing the result of `scores.sort()` | `sort()` changes the list and returns `None`; use `sorted()` for a copy. |
| Off by one | Expecting a slice to include the stop index | `items[1:4]` gives indexes 1, 2 and 3, not 4. |
| TypeError | Trying to change a tuple | Use a list if the values need to change. |
