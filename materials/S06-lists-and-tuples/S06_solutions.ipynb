### In-Class Exercise 3: Shopping Cart

```python
cart = []
for item in ["milk", "bread", "eggs"]:
    cart.append(item)
    print(cart, len(cart))
cart.insert(0, "coffee")
print(cart, len(cart))
cart.remove("bread")
print(cart, len(cart))
```

### In-Class Exercise 4: Middle Value

```python
def median(values):
    ordered = sorted(values)
    return ordered[len(ordered) // 2]

print(median([7, 1, 9, 3, 5]))
```

## Part B: Intermediate

### Exercise 3: Remove Duplicates

```python
def unique(values):
    result = []
    for v in values:
        if v not in result:
            result.append(v)
    return result

print(unique([3, 1, 3, 2, 1, 5, 2]))
```

### Exercise 4: Temperature Report

```python
temps = [21, 23, 25, 24, 28, 31, 30, 27, 22, 20, 19, 23, 26, 29]
week1, week2 = temps[:7], temps[7:]
print(f"Week 1 average: {sum(week1) / 7:.1f}")
print(f"Week 2 average: {sum(week2) / 7:.1f}")
print("Days above 25:", len([t for t in temps if t > 25]))
largest_rise = 0
for i in range(1, len(temps)):
    largest_rise = max(largest_rise, temps[i] - temps[i - 1])
print("Largest rise:", largest_rise)
```

## Part C: Challenge

### Exercise 5: Moving Average

A list of 7 values with a window of 3 gives 7 - 3 + 1 = 5 averages.

```python
def moving_average(values, window):
    result = []
    for i in range(len(values) - window + 1):
        part = values[i:i + window]
        result.append(round(sum(part) / window, 1))
    return result

print(moving_average([10, 12, 14, 13, 18, 20, 19], 3))
```

## Part D: Cumulative Project: Campus Bookshop

### Exercise 6: Bookshop Inventory

```python
titles = ["Python Basics", "Data Science Handbook", "Statistics Made Simple", "SQL in a Week"]
prices = [45.0, 62.5, 38.0, 29.9]
stock = [12, 3, 0, 7]

for t, p, s in zip(titles, prices, stock):
    print(f"{t:<25} {p:>6.2f} {s:>3}")

value = 0
for p, s in zip(prices, stock):
    value += p * s
print(f"Stock value: {value:.2f}")

out_of_stock = [t for t, s in zip(titles, stock) if s == 0]
low_stock = [t for t, s in zip(titles, stock) if s < 5]
print("Out of stock:", out_of_stock)
print("Low stock:", low_stock)

for price, title in sorted(zip(prices, titles), reverse=True):
    print(price, title)
```
