## 1. Creating and Reading Dictionaries

A dictionary is written with curly brackets. Each entry is `key: value`, and entries are separated by commas. Keys are usually strings and must be unique; values can be anything.

````python
prices = {"Espresso": 2.5, "Latte": 3.5, "Muffin": 3.0}

print(prices["Latte"])
print(len(prices))
print("Muffin" in prices)
print("Tea" in prices)
````

````python
prices = {"Espresso": 2.5, "Latte": 3.5, "Muffin": 3.0}
print(prices["Tea"])
````

````python
prices = {"Espresso": 2.5, "Latte": 3.5, "Muffin": 3.0}

print(prices.get("Tea"))
print(prices.get("Tea", 0))
print(prices.get("Latte", 0))
````

## 2. Changing Dictionaries

````python
stock = {"pen": 40, "notebook": 25}

stock["ruler"] = 15
stock["pen"] = 35
stock["notebook"] -= 5
del stock["ruler"]

print(stock)
print(stock.pop("pen"))
print(stock)
````

## 3. Looping over Dictionaries

````python
scores = {"Lina": 88, "Omar": 92, "Sara": 79}

for name in scores:
    print(name)

print(list(scores.values()))

for name, score in scores.items():
    print(f"{name} scored {score}")
````

````python
scores = {"Lina": 88, "Omar": 92, "Sara": 79}

best = max(scores, key=scores.get)
print("Best student:", best, scores[best])

passed = {name: s for name, s in scores.items() if s >= 80}
print(passed)
````

## 4. Counting with Dictionaries

Counting how often each value appears is one of the most useful things you can do with data: how many orders per product, how many students per city. A dictionary is perfect for this: the thing you count is the key, and the count is the value.

````python
orders = ["Latte", "Espresso", "Latte", "Muffin", "Latte", "Espresso"]

counts = {}
for item in orders:
    counts[item] = counts.get(item, 0) + 1

print(counts)
````

Python also has a ready-made counter in the `collections` module. You will learn about modules in Session 10, but here is a preview:

```python
from collections import Counter

```

## 5. Records: Dictionaries inside Lists

A dictionary can describe one **record**, such as one student or one product. A list of such dictionaries is a small table: each dictionary is a row and each key is a column. This is exactly the shape of data that pandas turns into a DataFrame in Session 16.

````python
students = [
    {"name": "Lina", "city": "Damascus", "gpa": 3.6},
    {"name": "Omar", "city": "Aleppo", "gpa": 3.1},
    {"name": "Sara", "city": "Damascus", "gpa": 3.9},
]

for s in students:
    print(f"{s['name']:<6} {s['city']:<10} {s['gpa']}")

in_damascus = [s["name"] for s in students if s["city"] == "Damascus"]
print(in_damascus)
````

````python
profile = {
    "name": "Lina",
    "skills": ["Python", "SQL"],
    "address": {"city": "Damascus", "country": "Syria"},
}

print(profile["skills"][0])
print(profile["address"]["city"])
````

## 6. Sets

A **set** is written with curly brackets like a dictionary, but holds single values instead of pairs. A set never contains duplicates, and its items have no order or index. Sets are ideal for removing duplicates and for comparing groups.

````python
cities = ["Damascus", "Aleppo", "Damascus", "Homs", "Aleppo"]
unique_cities = set(cities)
print(unique_cities)
print(len(unique_cities))

python_group = {"Lina", "Omar", "Sara", "Hadi"}
sql_group = {"Omar", "Hadi", "Maya"}
print(python_group & sql_group)
print(python_group | sql_group)
print(python_group - sql_group)
````

| Collection | Written as | Ordered | Changeable | Use it for |
| --- | --- | --- | --- | --- |
| List | `[1, 2, 3]` | Yes | Yes | A sequence of values, in order |
| Tuple | `(1, 2, 3)` | Yes | No | A fixed group of related values |
| Dictionary | `{"a": 1}` | Yes | Yes | Looking values up by a key |
| Set | `{1, 2, 3}` | No | Yes | Unique values and group comparisons |

## Common Mistakes

| Symptom | Mistake | Fix |
| --- | --- | --- |
| KeyError | Reading a key that does not exist | Check with `in`, or use `get(key, default)`. |
| Lost data | Assigning to an existing key by accident | Assignment replaces the old value; check with `in` before adding. |
| TypeError | Trying to index a set with `[0]` | Sets have no order; convert to a sorted list first. |
| Empty dict | Writing `{}` when you wanted an empty set | `{}` is an empty dictionary; use `set()` for an empty set. |
