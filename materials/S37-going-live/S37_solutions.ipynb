## Part A: Basic

### Exercise 1: Traceback Detective

```python
causes = {
    "ModuleNotFoundError sklearn": "scikit-learn missing from requirements.txt; add scikit-learn (the pip name, not sklearn)",
    "FileNotFoundError data/apartments.csv": "the data file is not in the repository at that path; upload it or fix the path",
    "KeyError district": "the DataFrame sent to the model has no district column; check the input column names",
    "InconsistentVersionWarning": "the model was saved with 1.5.2; pin scikit-learn==1.5.2 in requirements.txt",
}
for k, v in causes.items():
    print(f"{k}: {v}")
```

### Exercise 2: Pinned Requirements

```python
import importlib
def pinned(names):
    pip = {"sklearn": "scikit-learn"}
    lines = [f"{pip.get(n, n)}=={importlib.import_module(n).__version__}" for n in names]
    return "\n".join(lines + ["streamlit"])

print(pinned(["pandas", "sklearn", "joblib"]))
```

### Exercise 5: Model Information Page

Files in a `pages` folder next to `app.py` become extra pages, listed in the sidebar in alphabetical order.

```python
import json
from pathlib import Path
import streamlit as st

meta = json.loads((Path(__file__).parent.parent / "models" / "metadata.json").read_text())
st.title("About the model")
st.write(f"Trained on {meta['rows']} apartments on {meta['trained_on']}.")
c1, c2 = st.columns(2)
c1.metric("Typical error (MAE)", f"{meta['cv_mae_thousands']} thousand")
c2.metric("R2", meta["cv_r2"])
st.subheader("Features")
st.write(", ".join(meta["features"]))
st.subheader("Limits")
st.write("Practice data for one fictional city. Estimates are not valuations.")
```
