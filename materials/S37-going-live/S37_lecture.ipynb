## 1. Preparing the Repository

| File | Why |
| --- | --- |
| app.py | The app itself; the server runs `streamlit run app.py` |
| requirements.txt | The server installs exactly these packages |
| models/price_model.joblib | The fitted pipeline the app loads |
| data/ (optional) | Small data files the app reads |
| README.md | Not needed by the server, but by every visitor |

````python
packages = {"pandas": pd, "scikit-learn": sklearn, "joblib": joblib}
lines = [f"{name}=={module.__version__}" for name, module in packages.items()]
lines.append("streamlit")
print("\n".join(lines))
````

````python
def ready_to_deploy(folder):
    folder = Path(folder)
    needed = ["app.py", "requirements.txt", "models/price_model.joblib"]
    missing = [f for f in needed if not (folder / f).exists()]
    if missing:
        print("Missing:", ", ".join(missing))
    else:
        print("Ready to deploy")

Path("price-predictor/models").mkdir(parents=True, exist_ok=True)
Path("price-predictor/app.py").write_text("import streamlit as st\n")
ready_to_deploy("price-predictor")
````

| --- | --- |
| The last line | The type of error and the message |
| The file and line number | Where in your code it happened |
| The code line shown | The exact statement that failed |

Most deployment errors are one of three: a missing package, a wrong path, or a version mismatch.

## Common Mistakes

| Symptom | Mistake | Fix |
| --- | --- | --- |
| Private repository | Deploying from a private repository on a free plan | Make the repository public. |
| Unpinned versions | `scikit-learn` without a version | Pin the training version with `==`. |
| Absolute paths | `/content/models/price_model.joblib` | Use paths relative to the repository root. |
| Leaked key | An API key written in app.py | Use `st.secrets` and the Secrets settings. |
