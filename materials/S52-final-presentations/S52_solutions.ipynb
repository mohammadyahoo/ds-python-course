### In-Class Exercise 2: Your Key Chart

```python
exams["att_band"] = pd.cut(exams["attendance_pct"], [60, 80, 90, 100], labels=["below 80", "80-90", "90+"])
r = exams.groupby("att_band", observed=True)["passed"].mean() * 100
fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(r.index.astype(str), r, color=["#c9c5f7", "#c9c5f7", "#5b4fe0"])
for i, v in enumerate(r):
    ax.text(i, v + 2, f"{v:.0f}%", ha="center")
ax.set_title("Higher attendance, higher pass rate", loc="left")
ax.set_xlabel("Attendance (percent)")
ax.set_ylim(0, 100)
ax.set_yticks([])
ax.spines[["top", "right", "left"]].set_visible(False)
plt.show()
```
