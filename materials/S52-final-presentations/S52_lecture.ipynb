## Setup

Run this cell first. It loads the libraries and sets the address of the course datasets.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DATA_URL = "https://raw.githubusercontent.com/mohammadyahoo/ds-python-course/main/data/"
exams = pd.read_csv(DATA_URL + "exam_scores.csv")
```

## 1. A Five-Minute Structure

| Part | Time | Content |
| --- | --- | --- |
| 1. Problem | 40 seconds | Who has the problem and why it matters |
| 2. Data | 40 seconds | Source, size, target, one key chart |
| 3. Approach | 60 seconds | Baseline, models tried, how you evaluated |
| 4. Result | 60 seconds | Final score in plain words, the comparison chart |
| 5. Demo | 60 seconds | The live app with one realistic example |
| 6. Lessons | 20 seconds | One limit, one next step |

## 2. Charts for an Audience

```python
exams["study_band"] = pd.cut(exams["study_hours_week"], [0, 3, 6, 10, 30],
                             labels=["0-3", "3-6", "6-10", "10+"])
rates = exams.groupby("study_band", observed=True)["passed"].mean()
rates.plot(kind="bar")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
colors = ["#c9c5f7"] * 3 + ["#5b4fe0"]
ax.bar(rates.index.astype(str), rates * 100, color=colors)
for i, v in enumerate(rates * 100):
    ax.text(i, v + 2, f"{v:.0f}%", ha="center", fontsize=12)
ax.set_title("Students who study 10+ hours a week pass far more often", loc="left")
ax.set_xlabel("Study hours per week")
ax.set_ylim(0, 100)
ax.spines[["top", "right", "left"]].set_visible(False)
ax.set_yticks([])
plt.show()
```

## 4. Answering Questions

| Question | A strong answer includes |
| --- | --- |
| How accurate is it? | The test score in plain words, compared with the baseline |
| Why this model? | It matched or beat the others in cross-validation and is easier to explain |
| Can I trust it for my case? | Where it works well, where it struggles, and its data limits |
| Is it fair? | Error rates by group, and what you would check with more data |
| What would you do next? | One concrete improvement: more data, a new feature, monitoring |

"I do not know, but here is how I would find out" is a good answer.

## Common Mistakes

| Symptom | Mistake | Fix |
| --- | --- | --- |
| Running over | Spending three minutes on data cleaning | Rehearse with a timer; cut details. |
| Code on slides | Showing pipelines line by line | Show results; link to the repository. |
| Untested demo | Opening the app for the first time on stage | Wake it early and keep a backup. |
| Overclaiming | "The model is 100 percent reliable" | State the score, the baseline and the limits. |
