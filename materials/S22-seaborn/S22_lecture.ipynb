## Where We Are

In Session 21 you built charts with Matplotlib, one call at a time: group the data, draw the bars, add a colour per group by hand, add a legend. That gives full control but takes many lines.

**Seaborn** is built on top of Matplotlib and designed for statistics. You pass it a DataFrame and the names of the columns, and it groups, colours, adds legends and even calculates averages for you. Every Seaborn chart is still a Matplotlib chart, so `plt.title` and `plt.savefig` keep working.

## Setup

Run this cell first. Seaborn is imported as `sns`. `set_theme` gives every chart a clean style with a light grid. We use the bike rentals and a practice dataset of 500 students' exam results.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
DATA_URL = "https://raw.githubusercontent.com/mohammadyahoo/ds-python-course/main/data/"
bikes = pd.read_csv(DATA_URL + "bike_rentals.csv")
bikes["date"] = pd.to_datetime(bikes["date"])
exams = pd.read_csv(DATA_URL + "exam_scores.csv")
```

## 1. Distributions

```python
sns.histplot(data=exams, x="math", bins=25, kde=True)
plt.title("Math scores")
plt.show()
```

```python
sns.histplot(data=bikes, x="rentals", hue="weather",
             element="step", stat="density", common_norm=False)
plt.title("Daily rentals by weather")
plt.show()
```

## 2. Comparing Groups

```python
order = ["Winter", "Spring", "Summer", "Autumn"]
sns.boxplot(data=bikes, x="season", y="rentals", order=order)
plt.title("Daily rentals by season")
plt.show()
```

```python
sns.violinplot(data=bikes, x="weather", y="rentals",
               order=["Rain", "Cloudy", "Clear"], inner="quartile")
plt.title("Rentals by weather")
plt.show()
```

```python
sns.barplot(data=exams, x="class", y="math", order=list("ABCDE"))
plt.title("Average math score by class")
plt.show()
```

```python
sns.countplot(data=bikes, x="weather", hue="season",
              hue_order=["Winter", "Spring", "Summer", "Autumn"])
plt.title("Days of each weather type per season")
plt.show()
```

## 3. Relationships

```python
sns.scatterplot(data=bikes, x="temp_c", y="rentals",
                hue="weather", alpha=0.6, s=20)
plt.title("Rentals, temperature and weather")
plt.show()
```

```python
sns.regplot(data=exams, x="study_hours_week", y="math",
            scatter_kws={"s": 10, "alpha": 0.4})
plt.title("More study, higher scores")
plt.show()
```

## 4. Small Multiples

```python
sns.relplot(data=bikes, x="temp_c", y="rentals",
            col="weather", col_order=["Clear", "Cloudy", "Rain"],
            s=10, alpha=0.5, height=3.2)
plt.show()
```

## 5. Correlation Heatmaps

```python
cols = ["study_hours_week", "sleep_hours", "attendance_pct",
        "math", "physics", "english"]
corr = exams[cols].corr().round(2)

sns.heatmap(corr, annot=True, cmap="RdBu_r", vmin=-1, vmax=1)
plt.title("Correlation between exam variables")
plt.show()
```

| Question | Function | Key arguments |
| --- | --- | --- |
| How is one column spread? | `histplot`, `kdeplot` | `x`, `hue`, `bins` |
| How do groups compare? | `boxplot`, `violinplot`, `barplot` | `x` category, `y` number |
| How many rows per group? | `countplot` | `x`, `hue` |
| Do two numbers move together? | `scatterplot`, `regplot` | `x`, `y`, `hue` |
| The same chart per group? | `relplot`, `displot`, `catplot` | `col`, `row` |
| Which columns are related? | `heatmap` of `df.corr()` | `annot`, `cmap` |

## Common Mistakes

| Symptom | Mistake | Fix |
| --- | --- | --- |
| ValueError | Passing a column instead of its name: `x=df["a"]` with `data=` | With `data=df`, give names: `x="a"`. |
| Unordered | Seasons or weekdays in a random order | Pass `order=[...]` or `hue_order=[...]`. |
| Too many colours | Using `hue` with a column of 30 categories | Use `col=` for small multiples, or group first. |
| Misread | Reading a bar plot's line as data | The black line is a confidence interval, not a value. |
