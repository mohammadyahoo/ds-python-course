## Setup

Run this cell first. `sns.load_dataset("titanic")` downloads the passenger data that comes with Seaborn, so Colab needs an internet connection. The data describes 891 of the passengers.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
DATA_URL = "https://raw.githubusercontent.com/mohammadyahoo/ds-python-course/main/data/"
titanic = sns.load_dataset("titanic")
```

## 2. Data Quality

```python
titanic.isna().sum().sort_values(ascending=False).head(5)
```

## 3. Survival Overall and by Group

```python
print(round(titanic["survived"].mean(), 3))
print(titanic.groupby("sex")["survived"].mean().round(3))
```

```python
titanic.groupby("class")["survived"].agg(["count", "mean"]).round(3)
```

```python
sns.barplot(data=titanic, x="class", y="survived", hue="sex")
plt.title("Survival rate by class and sex")
plt.ylabel("Share who survived")
plt.show()
```

## 4. Combining Factors

```python
pd.crosstab(titanic["class"], titanic["sex"],
            values=titanic["survived"], aggfunc="mean").round(3)
```

```python
titanic.groupby("alone")["survived"].mean().round(3)
```

## 5. New Columns for New Questions

```python
bins = [0, 12, 18, 35, 60, 80]
labels = ["child", "teen", "young adult", "adult", "senior"]
titanic["age_group"] = pd.cut(titanic["age"], bins=bins, labels=labels)
titanic["family_size"] = titanic["sibsp"] + titanic["parch"] + 1

titanic.groupby("age_group", observed=True)["survived"].agg(["count", "mean"]).round(2)
```

```python
sns.histplot(data=titanic, x="age", hue="alive", bins=30, multiple="stack")
plt.title("Ages of survivors and non-survivors")
plt.show()
```

```python
titanic.groupby("class")["fare"].median()
```

## Common Mistakes

| Symptom | Mistake | Fix |
| --- | --- | --- |
| Misleading rate | Comparing counts of survivors instead of rates | Compare means of `survived`, because group sizes differ. |
| Hidden mix | Concluding from one factor alone | Check combinations with `crosstab` or `groupby` on two columns. |
| Silent gaps | Ignoring the 177 missing ages | Say how many rows each result covers. |
| Overclaiming | Writing that class caused survival | Describe patterns; offer explanations as possibilities. |
