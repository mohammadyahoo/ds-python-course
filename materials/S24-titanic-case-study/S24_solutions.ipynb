## Setup

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
DATA_URL = "https://raw.githubusercontent.com/mohammadyahoo/ds-python-course/main/data/"
titanic = sns.load_dataset("titanic")
```

## In-Class Exercises

### In-Class Exercise 1: Port of Boarding

```python
titanic.groupby("embark_town")["survived"].agg(["count", "mean"]).round(3)
```

### In-Class Exercise 2: Children by Class

```python
children = titanic[titanic["who"] == "child"]
children.groupby("class")["survived"].agg(["count", "mean"]).round(2)
```

### In-Class Exercise 3: Family Size

```python
titanic["family_size"] = titanic["sibsp"] + titanic["parch"] + 1
sns.barplot(data=titanic, x="family_size", y="survived")
plt.title("Survival rate by family size")
plt.show()
```

### In-Class Exercise 4: Fare Groups

```python
titanic["fare_group"] = pd.qcut(titanic["fare"], 4, labels=["low", "medium", "high", "very high"])
titanic.groupby("fare_group", observed=True)["survived"].mean().round(3)
```

## Part A: Basic

### Exercise 1: Basic Rates

```python
print(titanic["survived"].value_counts())
print(round(titanic["survived"].mean(), 3))
older = titanic[titanic["age"] > 60]
print(len(older), round(older["survived"].mean(), 3))
```

### Exercise 2: Fare Distribution

```python
sns.histplot(data=titanic, x="fare", bins=40)
plt.title("Fares")
plt.show()
sns.boxplot(data=titanic, x="class", y="fare")
plt.title("Fare by class")
plt.show()
print(round(titanic["fare"].mean(), 2), titanic["fare"].median())
# The median is more typical: a few very expensive tickets pull the mean up.
```

## Part B: Intermediate

### Exercise 3: A Pivot Table of Survival

```python
pd.pivot_table(titanic, index="class", columns="who", values="survived", aggfunc="mean", margins=True).round(2)
# Women and children survived far more often than men in every class.
# Even children in third class survived less often than children in first and second class.
```

### Exercise 4: Missing Ages

```python
titanic["age_known"] = titanic["age"].notna()
print(titanic.groupby("age_known")["survived"].mean().round(3))
print(pd.crosstab(titanic["age_known"], titanic["class"], normalize="index").round(2))
# Passengers without an age were more often in third class and survived less often, so age results may look slightly better than the truth.
```

## Part C: Challenge

### Exercise 5: Fill the Missing Ages

```python
filled = titanic["age"].fillna(titanic.groupby(["class", "who"])["age"].transform("median"))
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
sns.histplot(titanic["age"].dropna(), bins=30, ax=axes[0])
axes[0].set_title("Known ages")
sns.histplot(filled, bins=30, ax=axes[1])
axes[1].set_title("After filling by group")
plt.show()
# Group medians respect that children are young and first class passengers were older, so the filled values are more realistic.
```
