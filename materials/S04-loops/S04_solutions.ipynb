### In-Class Exercise 1: Even Numbers

### In-Class Exercise 2: Sum and Average

```python
n = 10
total = 0

for number in range(1, n + 1):
    total += number

print(f"Sum: {total}")
print(f"Average: {total / n:.2f}")
```

### In-Class Exercise 3: Counting a Letter

```python
sentence = "A data analyst asks a lot of questions about data"
count = 0

for character in sentence.lower():
    if character == "a":
        count += 1

print("The letter a appears", count, "times.")
```

### In-Class Exercise 4: Guess the Number

This run used the guesses 5, 9 and 7.

```python
secret = 7
guessed = False

for attempt in range(3):
    guess = int(input("Your guess: "))
    if guess == secret:
        print("Correct, well done.")
        guessed = True
        break
    elif guess > secret:
        print("Too high.")
    else:
        print("Too low.")

if not guessed:
    print(f"Out of attempts. The secret number was {secret}.")
```

## Part B: Intermediate

### Exercise 3: Sum of Digits

```python
number = 90715
digit_sum = 0

while number > 0:
    digit_sum += number % 10
    number = number // 10

print("Sum of digits:", digit_sum)
```

### Exercise 4: Prime Numbers

```python
prime_count = 0

for number in range(2, 51):
    is_prime = True
    for divisor in range(2, number):
        if number % divisor == 0:
            is_prime = False
            break
    if is_prime:
        print(number)
        prime_count += 1

print("Prime numbers from 2 to 50:", prime_count)
```

## Part C: Challenge

### Exercise 5: The Longest Collatz Sequence

```python
best_start = 1
most_steps = 0

for start in range(1, 21):
    number = start
    steps = 0
    while number != 1:
        if number % 2 == 0:
            number = number // 2
        else:
            number = number * 3 + 1
        steps += 1
    if steps > most_steps:
        most_steps = steps
        best_start = start

print(f"Starting number {best_start} needs the most steps: {most_steps}.")
```

Both 18 and 19 need 20 steps. The condition `steps > most_steps` only replaces the best value when a new one is strictly larger, so the first of the two, 18, is kept. With `>=`, the answer would be 19.

## Part D: Cumulative Project: Campus Bookshop

### Exercise 6: One Week of Sales

```python
price = 45.0
stock = 15

total_sold = 0
revenue = 0
shortage_days = 0
restock_orders = 0

for day in range(1, 8):
    wanted = 4 + (day * 3) % 7

    if wanted <= stock:
        sold = wanted
    else:
        sold = stock
        shortage_days += 1

    stock -= sold
    total_sold += sold
    revenue += sold * price

    print(f"Day {day}: wanted {wanted}, sold {sold}, stock at close {stock}")

    if stock < 5:
        stock += 20
        restock_orders += 1

print("-" * 30)
print(f"Copies sold: {total_sold}")
print(f"Revenue: {revenue:.2f}")
print(f"Shortage days: {shortage_days}")
print(f"Restock orders: {restock_orders}")
```
