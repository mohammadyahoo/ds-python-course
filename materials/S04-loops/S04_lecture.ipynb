## 1. The for Loop and range()

A `for` loop repeats a block once for each value in a sequence. `range(5)` produces the numbers 0, 1, 2, 3 and 4.

```python
for variable in sequence:
    code that runs once for each value
```

As with `if`, the line ends with a colon and the block is indented.

````python
for number in range(5):
    print("Repetition number", number)

print("Loop finished.")
````

On each repetition, the variable `number` takes the next value from the sequence. The last line is not indented, so it runs once, after the loop ends.

`range()` can take up to three values:

| Call | Values produced | Meaning |
| --- | --- | --- |
| `range(5)` | 0, 1, 2, 3, 4 | From 0 up to, but not including, 5 |
| `range(1, 6)` | 1, 2, 3, 4, 5 | From 1 up to, but not including, 6 |
| `range(0, 20, 5)` | 0, 5, 10, 15 | From 0 in steps of 5 |
| `range(10, 0, -2)` | 10, 8, 6, 4, 2 | Counting down in steps of 2 |

The stop value is **never included**. This is the most common source of off-by-one errors.

````python
for year in range(2024, 2029):
    is_leap = (year % 4 == 0 and year % 100 != 0) or year % 400 == 0
    print(year, is_leap)
````

Three lines now do the work of any number of repeated lines from Session 03.

## 2. Looping Over a String

A string is a sequence of characters, so a `for` loop can visit each character in turn.

````python
word = "Python"

for letter in word:
    print(letter)
````

## 3. The Accumulator Pattern

Many questions about data are answered by building up a result while looping: a total, a count, or the largest value. This is called the **accumulator pattern**, and it always has three parts:

1. **Before the loop**, create a variable with a starting value.
2. **Inside the loop**, update it.
3. **After the loop**, use the result.

### A total

````python
total = 0

for number in range(1, 101):
    total += number

print("Sum of 1 to 100:", total)
````

### A count

Combine the loop with an `if` statement to count only the values that meet a condition. Here we count the vowels in a sentence, using `in` from Session 03.

````python
sentence = "Data science turns raw data into decisions"
vowel_count = 0

for character in sentence.lower():
    if character in "aeiou":
        vowel_count += 1

print("Vowels:", vowel_count)
````

`sentence.lower()` gives a lowercase copy of the sentence, so capital letters are counted too. You will learn more string tools in Session 07.

### A maximum

To find the largest value, start with the first candidate and replace it whenever a larger one appears. Here we find which whole number from 1 to 10 gives the largest value of the expression `n * (12 - n)`.

````python
best_n = 1
best_value = 1 * (12 - 1)

for n in range(1, 11):
    value = n * (12 - n)
    if value > best_value:
        best_value = value
        best_n = n

print(f"The largest value is {best_value}, reached when n is {best_n}.")
````

## 4. The while Loop

A `for` loop is the right choice when you know how many times to repeat. A `while` loop repeats **as long as a condition is true**, which is useful when you do not know the number of repetitions in advance.

````python
balance = 1000
years = 0

while balance < 2000:
    balance = balance * 1.08
    years += 1

print(f"At 8 percent per year, the balance doubles after {years} years.")
````

The condition is checked before each repetition. Something inside the loop **must** eventually make it false, otherwise the loop never stops. This is called an **infinite loop**. If it happens in Colab, click the stop button next to the cell.

A common use of `while` is to keep asking for input until it is valid. Run the cell and try typing a mark above 100 first.

````python
mark = int(input("Enter a mark from 0 to 100: "))

while mark < 0 or mark > 100:
    print("That is not a valid mark.")
    mark = int(input("Enter a mark from 0 to 100: "))

print("Mark recorded:", mark)
````

## 5. Controlling a Loop with break and continue

1. `break` stops the loop immediately, even if values remain.
2. `continue` skips the rest of the current repetition and moves to the next value.

`break` is useful for searching: stop as soon as you find what you need.

````python
for number in range(100, 1000):
    if number % 7 == 0 and number % 11 == 0:
        print("First number above 100 divisible by 7 and 11:", number)
        break
````

`continue` is useful for skipping values you do not want to process.

````python
for number in range(1, 11):
    if number % 3 == 0:
        continue
    print(number)
````

## 6. Loops Inside Loops

A loop can contain another loop. For each repetition of the outer loop, the inner loop runs completely. A multiplication table is a classic example.

````python
for row in range(1, 4):
    for column in range(1, 6):
        print(f"{row} x {column} = {row * column}")
    print("-" * 15)
````

## 7. Common Mistakes

1. **Off-by-one errors.** `range(1, 10)` stops at 9. Use `range(1, 11)` to include 10.
2. **Infinite `while` loops.** The variable in the condition must change inside the loop.
3. **Resetting an accumulator inside the loop.** `total = 0` belongs before the loop, not inside it.
4. **Wrong indentation of a `print`.** Indented, it runs on every repetition. Not indented, it runs once after the loop.

Compare the two cells below. They differ only in the indentation of the last line.

````python
total = 0
for number in range(1, 4):
    total += number
    print("Total so far:", total)
````

````python
total = 0
for number in range(1, 4):
    total += number
print("Final total:", total)
````

## 8. Worked Example: Saving for a Laptop

A student wants to buy a laptop that costs 3200. She already has 500, saves 250 at the end of every month, and her savings account adds 1 percent interest to the balance at the start of each month. How many months does she need, and what does the balance look like along the way?

We do not know the number of months in advance, so this is a job for `while`.

**Step 1.** Store the facts in variables.

````python
target = 3200
balance = 500
monthly_saving = 250
monthly_rate = 0.01
month = 0
````

**Step 2.** Repeat month by month until the balance reaches the target. Each month, add interest, then add the saving, then print a line of the table.

````python
while balance < target:
    month += 1
    interest = balance * monthly_rate
    balance = balance + interest + monthly_saving
    print(f"Month {month}: interest {interest:.2f}, balance {balance:.2f}")
````

**Step 3.** Report the answer.

````python
print(f"The laptop is affordable after {month} months, with {balance - target:.2f} left over.")
````

Try changing `monthly_saving` to 300 and running Steps 1 to 3 again. Because every value is stored in a variable, the whole table updates.
