```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.dummy import DummyClassifier

DATA_URL = "https://raw.githubusercontent.com/mohammadyahoo/ds-python-course/main/data/"
exams = pd.read_csv(DATA_URL + "exam_scores.csv")
```

## 3. Profiling a New Dataset

```python
def quick_profile(df):
    profile = pd.DataFrame({
        "type": df.dtypes.astype(str),
        "missing": df.isna().sum(),
        "unique": df.nunique(),
        "example": df.iloc[0],
    })
    print(f"{df.shape[0]} rows, {df.shape[1]} columns, "
          f"{df.duplicated().sum()} duplicate rows")
    return profile
```

```python
print(quick_profile(exams))
```

```python
print(exams["passed"].value_counts(normalize=True).round(2))
print(exams.groupby("class")["passed"].mean().round(2))
```

## 4. Leakage and Baselines

```python
print(exams[["math", "physics", "english", "passed"]].corr()["passed"].round(2))
```

```python
X = exams[["study_hours_week", "sleep_hours", "attendance_pct"]]
y = exams["passed"]

dummy = DummyClassifier(strategy="most_frequent")
logistic = make_pipeline(StandardScaler(), LogisticRegression())
for name, model in [("always 'fail'", dummy), ("logistic", logistic)]:
    print(name, cross_val_score(model, X, y, cv=5).mean().round(3))
```

## Common Mistakes

| Symptom | Mistake | Fix |
| --- | --- | --- |
| Too big | Planning five models, three apps and a database | Finish version 1 first, then improve. |
| No target | A dataset with no column to predict | Pick a question the data can answer. |
| Leaky features | Using exam marks to predict passing | Use only inputs known at prediction time. |
| Late deploy | Deploying the night before the presentation | Deploy version 1 early and update it. |
