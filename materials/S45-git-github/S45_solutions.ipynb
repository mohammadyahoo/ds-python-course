## Part A: Basic

### Exercise 1: Git Vocabulary

```python
commands = {
    "git init": "turn a folder into a repository",
    "git status": "show which files changed or are staged",
    "git add": "stage changes for the next commit",
    "git commit": "save a snapshot with a message",
    "git log": "list the commit history",
    "git diff": "show the exact line changes",
    "git switch -c": "create a new branch and move to it",
    "git clone": "copy a repository from GitHub",
}
for k, v in commands.items():
    print(f"{k:15} {v}")
```

### Exercise 2: Rewrite the Messages

### Exercise 4: Undo a Mistake

`git restore` brings back the last committed version of a file. It only works for changes that are not committed yet.

```python
!echo "oops" > notes.txt
!git diff
!git restore notes.txt
!git status
!cat notes.txt
```

## Part C: Challenge

### Exercise 5: Write a README with Python

```python
results = {"Test AUC": 0.80, "Recall at 0.3": 0.78, "Precision at 0.3": 0.53}
lines = ["# Customer Churn Model", "",
         "Predicts which telecom customers are likely to leave, so the retention team can call them first.", "",
         "## Results", "", "| Metric | Value |", "| --- | --- |"]
lines += [f"| {k} | {v:.2f} |" for k, v in results.items()]
lines += ["", "## How to run", "", "Open `churn_model.ipynb` in Google Colab and run all cells.", "",
          "## Data", "", "Practice data: 2,000 fictional customers from the course data repository."]
with open("README.md", "w") as f:
    f.write("\n".join(lines) + "\n")
print(open("README.md").read())
```
