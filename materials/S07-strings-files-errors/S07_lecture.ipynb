## 2. Cleaning Text

String methods never change the original string, because strings are immutable. They return a **new** string, so you store the result: `name = name.strip()`. Methods can be **chained**: each one works on the result of the one before.

```python
raw = "   lina HADDAD  "

print(repr(raw.strip()))
print(raw.strip().lower())
print(raw.strip().upper())
print(raw.strip().title())
print("2,500,000".replace(",", ""))
```

```python
cities = ["Damascus", "damascus ", "DAMASCUS", " Aleppo", "aleppo"]

clean = [c.strip().title() for c in cities]
print(clean)
print(set(clean))
```

## 3. Searching Text

```python
email = "lina.haddad@university.edu"

print("@" in email)
print(email.endswith(".edu"))
print(email.startswith("omar"))
print(email.find("@"))
print(email.count("a"))
print(email[email.find("@") + 1:])
```

| Method | True when the string | Example that is True |
| --- | --- | --- |
| `isdigit()` | has only digits | `"2025".isdigit()` |
| `isalpha()` | has only letters | `"Lina".isalpha()` |
| `isupper()` | has no lowercase letters | `"SQL".isupper()` |
| `startswith(x)` | begins with `x` | `"DS-01".startswith("DS")` |
| `endswith(x)` | finishes with `x` | `"data.csv".endswith(".csv")` |

## 4. Splitting and Joining

`split()` breaks a string into a list. With no argument it splits on any spaces; with an argument such as `","` it splits on that character. `join()` does the opposite: the string before the dot is the glue placed between the items.

```python
sentence = "Python makes data science fun"
words = sentence.split()
print(words)
print(len(words), "words")

row = "Lina,Damascus,3.6"
name, city, gpa = row.split(",")
print(name, city, float(gpa) + 0.1)
```

```python
words = ["Python", "makes", "data", "science", "fun"]
print(" ".join(words))
print(" | ".join(words))
print("-".join(["2025", "03", "14"]))
```

## 5. Formatting Output

You met `:.2f` in Session 02. A format code can also set a **width** and an **alignment**, so values line up in columns: `<` aligns left, `>` aligns right and `^` centres. A comma adds thousands separators, and `%` shows a fraction as a percentage.

```python
items = [("Latte", 3.5, 128),
         ("Croissant", 2.8, 64),
         ("Orange Juice", 3.8, 9)]

print(f"{'Item':<14}{'Price':>7}{'Sold':>6}")
for name, price, sold in items:
    print(f"{name:<14}{price:>7.2f}{sold:>6}")

print(f"{1234567.891:,.2f}")
print(f"{0.1834:.1%}")
```

| Code | Example | Result |
| --- | --- | --- |
| `:.2f` | `f"{3.14159:.2f}"` | `3.14` |
| `:>8` | `f"{42:>8}"` | `      42` |
| `:<8` | `f"{'ab':<8}|"` | `ab      |` |
| `:^9` | `f"{'mid':^9}"` | `   mid   ` |
| `:,` | `f"{2500000:,}"` | `2,500,000` |
| `:.0%` | `f"{0.75:.0%}"` | `75%` |

## 6. From Messy Text to Clean Data

```python
records = ["  Lina Haddad ; DAMASCUS ; 3.62 ",
           "omar saleh;aleppo;3.1",
           "Sara Nassar ; Homs ;3.95"]

students = []
for line in records:
    name, city, gpa = line.split(";")
    students.append({"name": name.strip().title(),
                     "city": city.strip().title(),
                     "gpa": float(gpa)})

for s in students:
    print(s)
```

## 7. Writing and Reading Text Files

`open(name, mode)` connects your program to a file. The mode `"w"` **writes** a new file (replacing any old one), `"a"` **appends** to the end, and `"r"` **reads**, which is the default.

Always open files with a `with` block. When the block ends, Python closes the file automatically, even if an error happens inside it.

In Colab, files are saved in the notebook's temporary folder. Click the folder icon on the left to see them. They are deleted when the runtime restarts, so download anything you want to keep.

```python
with open("notes.txt", "w") as f:
    f.write("Session 07: Files\n")
    f.write("Files keep data after the program ends.\n")

with open("notes.txt") as f:
    text = f.read()

print(text)
print(len(text), "characters")
```

```python
with open("notes.txt", "a") as f:
    f.write("Appending adds lines to the end.\n")

with open("notes.txt") as f:
    for number, line in enumerate(f, start=1):
        print(number, line.strip())
```

| Mode | Meaning | If the file exists | If it does not exist |
| --- | --- | --- | --- |
| `"r"` | Read (the default) | Reads it | `FileNotFoundError` |
| `"w"` | Write | Replaces everything | Creates it |
| `"a"` | Append | Adds to the end | Creates it |

## 8. CSV Files

A **CSV** file stores a table as text: one row per line, with the values separated by commas. The first line usually holds the column names. Every spreadsheet program can save and open CSV files, and pandas reads them in one line, as you will see in Session 11.

Here we write a small CSV file and read it with what you already know: `split(",")`.

```python
rows = [
    "item,category,price,sold",
    "Latte,Coffee,3.5,128",
    "Croissant,Pastry,2.8,64",
    "Green Tea,Tea,2.5,41",
]
with open("menu.csv", "w") as f:
    f.write("\n".join(rows) + "\n")
```

```python
with open("menu.csv") as f:
    header = f.readline().strip().split(",")
    for line in f:
        item, category, price, sold = line.strip().split(",")
        print(f"{item:<10} revenue {float(price) * int(sold):.2f}")

print(header)
```

## 9. Reading Error Messages

When Python cannot run a line, it stops and shows an **error message**, also called a traceback. Read it from the **bottom**: the last line names the type of error and gives a short description. The lines above show where it happened.

You have already met several errors. Here are the ones you will see most in data work.

| Error | Typical cause | Example |
| --- | --- | --- |
| `NameError` | A name that was never defined, often a typo | `print(totl)` |
| `TypeError` | An operation on the wrong type | `"5" + 3` |
| `ValueError` | The right type but an unusable value | `int("seven")` |
| `IndexError` | A list index that does not exist | `[1, 2][5]` |
| `KeyError` | A dictionary key that does not exist | `{"a": 1}["b"]` |
| `ZeroDivisionError` | Dividing by zero | `10 / 0` |
| `FileNotFoundError` | A wrong file name or folder | `open("data.cvs")` |

## 10. Handling Errors with try and except

Code that might fail goes inside a `try` block. If an error happens, Python jumps to the matching `except` block instead of stopping. Always name the error you expect, such as `except ValueError:`, so that other, unexpected errors are still reported.

```python
values = ["12.5", "8", "seven", "4.25"]
total = 0
for v in values:
    try:
        total += float(v)
    except ValueError:
        print(f"Skipping bad value: {v!r}")

print("Total:", total)
```

```python
def read_numbers(filename):
    try:
        with open(filename) as f:
            return [float(line) for line in f if line.strip()]
    except FileNotFoundError:
        print(f"File {filename} not found, using an empty list")
        return []

print(read_numbers("missing.txt"))
```

## 11. Raising Your Own Errors

Sometimes the data is valid Python but makes no sense: a negative price, an age of 230, an empty list where values are needed. Use `raise` to stop with a clear message. It is much better to fail loudly with a clear reason than to continue and produce a wrong answer.

```python
def apply_discount(price, rate):
    if not 0 <= rate <= 1:
        raise ValueError(f"rate must be between 0 and 1, got {rate}")
    return price * (1 - rate)

print(apply_discount(100, 0.2))
print(apply_discount(100, 20))
```

## Common Mistakes

| Symptom | Mistake | Fix |
| --- | --- | --- |
| No change | Calling `name.strip()` without storing the result | Strings never change; write `name = name.strip()`. |
| ValueError | Unpacking a split into the wrong number of names | Check the separator and the number of fields per line. |
| Data lost | Opening an existing file with mode `"w"` | `"w"` replaces the file; use `"a"` to add to it. |
| FileNotFoundError | A typo in the file name or extension | Check the Files panel; names are case sensitive. |
