## Setup

Run this cell first. It loads the libraries and sets the address of the course datasets.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DATA_URL = "https://raw.githubusercontent.com/mohammadyahoo/ds-python-course/main/data/"
bikes = pd.read_csv(DATA_URL + "bike_rentals.csv")
bikes["date"] = pd.to_datetime(bikes["date"])
import seaborn as sns

sns.set_theme(style="whitegrid")
exams = pd.read_csv(DATA_URL + "exam_scores.csv")
```

## 1. Your First Chart

Matplotlib builds a chart step by step: draw the data, then add a title and axis labels. In Colab the chart appears below the cell. `plt.show()` is not needed in Colab, but it is good practice: it tells Python the chart is finished.

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
rentals = [2410, 2650, 3480, 4320, 5050, 5560]

plt.plot(months, rentals, marker="o")
plt.title("Average daily rentals, first half of 2023")
plt.xlabel("Month")
plt.ylabel("Rentals per day")
plt.show()
```

## 2. Line Charts for Change over Time

```python
plt.figure(figsize=(9, 4))
plt.plot(bikes["date"], bikes["rentals"], linewidth=0.8)
plt.title("Daily bike rentals, 2023 to 2024")
plt.ylabel("Rentals")
plt.show()
```

## 3. Bar Charts to Compare Groups

```python
by_weather = bikes.groupby("weather")["rentals"].mean()
by_weather = by_weather.sort_values()

plt.bar(by_weather.index, by_weather.values,
        color=["tab:gray", "tab:blue", "tab:orange"])
plt.title("Average rentals by weather")
plt.ylabel("Rentals per day")
plt.show()
```

## 4. Scatter Plots for Relationships

```python
colors = {"Clear": "tab:orange", "Cloudy": "tab:gray",
          "Rain": "tab:blue"}
for weather, part in bikes.groupby("weather"):
    plt.scatter(part["temp_c"], part["rentals"], s=8, alpha=0.5,
                color=colors[weather], label=weather)

plt.title("Rentals, temperature and weather")
plt.xlabel("Temperature (C)")
plt.ylabel("Rentals")
plt.legend()
plt.show()
```

## 5. Histograms for Distributions

```python
plt.hist(bikes["rentals"], bins=30,
         color="tab:purple", edgecolor="white")
plt.title("How many rentals on a typical day?")
plt.xlabel("Rentals per day")
plt.ylabel("Number of days")
plt.show()
```

| Question | Chart | Matplotlib |
| --- | --- | --- |
| How does it change over time? | Line | `plt.plot` |
| How do groups compare? | Bar | `plt.bar`, `plt.barh` |
| Do two measures move together? | Scatter | `plt.scatter` |
| How are the values spread? | Histogram | `plt.hist` |
| What share does each part have? | Bar of shares (avoid pies) | `plt.bar` |

## 6. Several Charts in One Figure

`plt.subplots(rows, columns)` creates a figure and a grid of **axes**, one per chart. Each axes object has the same drawing methods as `plt`, and its labels are set with `set_title`, `set_xlabel` and so on. This style is called the object-oriented interface, and you will see it in most examples online.

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].hist(bikes["temp_c"], bins=25, color="tab:orange")
axes[0].set_title("Temperature")
axes[0].set_xlabel("C")

season_avg = bikes.groupby("season")["rentals"].mean().sort_values()
axes[1].bar(season_avg.index, season_avg.values, color="tab:green")
axes[1].set_title("Average rentals by season")

fig.tight_layout()
plt.show()
```

## 7. Distributions

```python
sns.histplot(data=bikes, x="rentals", hue="weather",
             element="step", stat="density", common_norm=False)
plt.title("Daily rentals by weather")
plt.show()
```

## 8. Comparing Groups

```python
order = ["Winter", "Spring", "Summer", "Autumn"]
sns.boxplot(data=bikes, x="season", y="rentals", order=order)
plt.title("Daily rentals by season")
plt.show()
```

```python
sns.barplot(data=exams, x="class", y="math", order=list("ABCDE"))
plt.title("Average math score by class")
plt.show()
```

```python
sns.countplot(data=bikes, x="weather", hue="season",
              hue_order=["Winter", "Spring", "Summer", "Autumn"])
plt.title("Days of each weather type per season")
plt.show()
```

## 9. Relationships

```python
sns.regplot(data=exams, x="study_hours_week", y="math",
            scatter_kws={"s": 10, "alpha": 0.4})
plt.title("More study, higher scores")
plt.show()
```

## 10. Correlation Heatmaps

```python
cols = ["study_hours_week", "sleep_hours", "attendance_pct",
        "math", "physics", "english"]
corr = exams[cols].corr().round(2)

sns.heatmap(corr, annot=True, cmap="RdBu_r", vmin=-1, vmax=1)
plt.title("Correlation between exam variables")
plt.show()
```

| Question | Function | Key arguments |
| --- | --- | --- |
| How is one column spread? | `histplot`, `kdeplot` | `x`, `hue`, `bins` |
| How do groups compare? | `boxplot`, `violinplot`, `barplot` | `x` category, `y` number |
| How many rows per group? | `countplot` | `x`, `hue` |
| Do two numbers move together? | `scatterplot`, `regplot` | `x`, `y`, `hue` |
| The same chart per group? | `relplot`, `displot`, `catplot` | `col`, `row` |
| Which columns are related? | `heatmap` of `df.corr()` | `annot`, `cmap` |

## Common Mistakes

| Symptom | Mistake | Fix |
| --- | --- | --- |
| Unreadable | A chart with no title or axis labels | Always add a title and label both axes with units. |
| Misleading | A bar chart whose y axis does not start at zero | Bars must start at zero; lines may zoom in. |
| ValueError | Passing a column instead of its name: `x=df["a"]` with `data=` | With `data=df`, give names: `x="a"`. |
| Too many colours | Using `hue` with a column of 30 categories | Use `col=` for small multiples, or group first. |
