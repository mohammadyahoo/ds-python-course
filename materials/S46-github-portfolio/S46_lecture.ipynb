## 1. Project Structure

| Path | What goes there |
| --- | --- |
| README.md | The project's front page: problem, result, how to run |
| requirements.txt | The Python packages and versions needed |
| .gitignore | Files Git should never track |
| notebooks/ | Analysis and modelling notebooks, numbered in order |
| data/ | Small data files, or a note on where to get the data |
| models/ | Saved model files (Session 47) |
| app.py | The web app (Session 48) |

```python
project = Path("price-predictor")
for folder in ["notebooks", "data", "models"]:
    (project / folder).mkdir(parents=True, exist_ok=True)

for path in sorted(project.rglob("*")):
    print(path)
```

## 2. The README

```python
readme = '''# Price Predictor

Estimates the price of an apartment from its size, age, location and features,
so buyers and sellers can check whether an asking price is reasonable.

## Result
A linear regression pipeline predicts prices with a typical error of about
10,000 (MAE), explaining about 75 percent of the variation on unseen apartments.

## How to run
Open `notebooks/01_model.ipynb` in Google Colab and run all cells.
'''
(project / "README.md").write_text(readme)
print((project / "README.md").read_text()[:120])
```

| You type | You get |
| --- | --- |
| # Title | A large heading |
| ## Section | A section heading |
| **bold** | Bold text |
| - item | A bullet point |
| [text](https://link) | A link |
| ![chart](images/chart.png) | An image from the repository |
| | a | b | | A table row |

## 3. Reproducible Projects

```python
print("Python", sys.version.split()[0])
print("pandas", pd.__version__)
print("scikit-learn", sklearn.__version__)
```

```python
requirements = f'''pandas=={pd.__version__}
scikit-learn=={sklearn.__version__}
'''
(project / "requirements.txt").write_text(requirements)
print((project / "requirements.txt").read_text())
```

```python
ignore = '''# Python
__pycache__/
.ipynb_checkpoints/

# Secrets and local settings
.env
secrets.toml

# Large or private data
data/raw/
'''
(project / ".gitignore").write_text(ignore)
print(len(ignore.splitlines()), "lines written")
```

## 4. Your Profile

| Check | Why it matters |
| --- | --- |
| Clear name and one-line description | Shown on your profile and in search |
| README with result near the top | Most visitors read nothing else |
| Notebooks run from top to bottom | Broken notebooks lose trust |
| requirements.txt | Others can reproduce the work |
| No secrets, no huge files | Safety and a fast download |
| Topics added (python, machine-learning) | Helps people find the project |

## Common Mistakes

| Symptom | Mistake | Fix |
| --- | --- | --- |
| Wall of code | README that only says "my project" | Lead with the problem and the result. |
| Untitled0.ipynb | Default notebook names | Name files by purpose: `01_cleaning.ipynb`. |
| No versions | requirements.txt with names but no versions | Pin versions with `==`. |
| Everything public | Committing personal or client data | Use practice or public data, and say so. |
