### In-Class Exercise 4: Check a Repository

```python
def check_repo(folder):
    folder = Path(folder)
    for name in ["README.md", "requirements.txt", ".gitignore"]:
        print(f"{name:18}", "ok" if (folder / name).exists() else "MISSING")
    big = [p for p in folder.rglob("*") if p.is_file() and p.stat().st_size > 50 * 1024 ** 2]
    print("Large files:", big if big else "none")

Path("price-predictor").mkdir(exist_ok=True)
check_repo("price-predictor")
```

### Exercise 1: Folder Tree

```python
def show_tree(folder):
    folder = Path(folder)
    print(folder.name + "/")
    for p in sorted(folder.rglob("*")):
        depth = len(p.relative_to(folder).parts)
        print("    " * depth + p.name + ("/" if p.is_dir() else ""))

demo = Path("demo-project")
(demo / "notebooks").mkdir(parents=True, exist_ok=True)
(demo / "data").mkdir(exist_ok=True)
(demo / "README.md").write_text("# Demo")
(demo / "notebooks" / "01_explore.ipynb").write_text("{}")
show_tree(demo)
```

### Exercise 3: requirements.txt from Imports

```python
import importlib
def requirements(imports):
    pip_names = {"sklearn": "scikit-learn"}
    lines = []
    for name in imports:
        module = importlib.import_module(name)
        lines.append(f"{pip_names.get(name, name)}=={module.__version__}")
    return "\n".join(lines)

print(requirements(["pandas", "sklearn", "numpy"]))
```

### Exercise 4: A Results Table for the README

`to_markdown` needs the small `tabulate` package. The fallback builds the same table by hand, so the cell works either way.

```python
results = pd.DataFrame({"Model": ["Mean baseline", "Linear regression", "Gradient boosting"],
                        "MAE (thousands)": [32.0, 10.1, 10.9], "R2": [0.00, 0.75, 0.73]})
try:
    print(results.to_markdown(index=False))
except ImportError:
    print("| " + " | ".join(results.columns) + " |")
    print("| " + " | ".join("---" for _ in results.columns) + " |")
    for row in results.itertuples(index=False):
        print("| " + " | ".join(str(v) for v in row) + " |")
```

### Exercise 5: Repository Score

```python
def score_repo(folder):
    folder = Path(folder)
    score, missing = 0, []
    for name in ["README.md", "requirements.txt", ".gitignore"]:
        if (folder / name).exists():
            score += 2
        else:
            missing.append(name)
    readme = folder / "README.md"
    if readme.exists() and len(readme.read_text()) > 500:
        score += 2
    else:
        missing.append("README longer than 500 characters")
    if not any(p.is_file() and p.stat().st_size > 50 * 1024 ** 2 for p in folder.rglob("*")):
        score += 2
    else:
        missing.append("no files above 50 MB")
    return score, missing

Path("empty-project").mkdir(exist_ok=True)
for f in ["price-predictor", "empty-project"]:
    Path(f).mkdir(exist_ok=True)
    print(f, score_repo(f))
```

### Exercise 6: Price Predictor App: Step 2, A Professional Repository

An empty file called `.gitkeep` is a common trick to keep an otherwise empty folder in a repository.

```python
# Generate the files in Colab, download them, and upload them to GitHub (Add file, Upload files).
project = Path("price-predictor")
for folder in ["notebooks", "data", "models"]:
    (project / folder).mkdir(parents=True, exist_ok=True)
    (project / folder / ".gitkeep").write_text("")
(project / "requirements.txt").write_text(f"pandas=={pd.__version__}\nscikit-learn=={sklearn.__version__}\n")
(project / ".gitignore").write_text("__pycache__/\n.ipynb_checkpoints/\n.env\nsecrets.toml\n")
for p in sorted(project.rglob("*")):
    print(p)
```
