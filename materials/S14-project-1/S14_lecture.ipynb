## 3. Building the Program

Build the program in small steps and run each step before writing the next. Start with the data classes, then the file functions, then the menu. Below is the full bookshop program, split into cells in that order.

```python
class Book:
    TAX_RATE = 0.15

    def __init__(self, title, price, stock, category):
        self.title = title
        self.price = float(price)
        self.stock = int(stock)
        self.category = category

    def sell(self, quantity):
        if quantity <= 0 or quantity > self.stock:
            raise ValueError(f"cannot sell {quantity}; {self.stock} in stock")
        rate = 0.10 if quantity >= 10 else 0
        self.stock -= quantity
        return round(self.price * quantity * (1 - rate) * (1 + Book.TAX_RATE), 2)

    def __str__(self):
        return f"{self.title:<26}{self.category:<13}{self.price:>7.2f}{self.stock:>6}"
```

```python
def save_catalogue(books, filename):
    with open(filename, "w") as f:
        f.write("title,price,stock,category\n")
        for b in books.values():
            f.write(f"{b.title},{b.price},{b.stock},{b.category}\n")

def load_catalogue(filename):
    books = {}
    try:
        with open(filename) as f:
            f.readline()
            for line in f:
                title, price, stock, category = line.strip().split(",")
                books[title] = Book(title, price, stock, category)
    except FileNotFoundError:
        print("No catalogue found, starting empty")
    return books

starter = {t: Book(t, p, s, c) for t, p, s, c in [
    ("Python Basics", 45.0, 12, "Programming"), ("SQL in a Week", 29.9, 7, "Programming"),
    ("Data Science Handbook", 62.5, 3, "Data"), ("Statistics Made Simple", 38.0, 0, "Data")]}
save_catalogue(starter, "catalogue.csv")
print(len(load_catalogue("catalogue.csv")), "books loaded")
```

```python
def ask_int(prompt):
    while True:
        try:
            return int(input(prompt))
        except ValueError:
            print("Please type a whole number.")

def show(books):
    print(f"{'Title':<26}{'Category':<13}{'Price':>7}{'Stock':>6}")
    for b in sorted(books.values(), key=lambda b: b.title):
        print(b)

def search(books, word):
    return [b for b in books.values() if word.lower() in b.title.lower()]
```

```python
def main():
    books = load_catalogue("catalogue.csv")
    revenue = 0
    menu = "1 List  2 Search  3 Sell  4 Report  5 Save and quit"
    while True:
        print("\n" + menu)
        choice = input("Choose: ").strip()
        if choice == "1":
            show(books)
        elif choice == "2":
            for b in search(books, input("Search for: ")):
                print(b)
        elif choice == "3":
            book = books.get(input("Title: ").strip())
            if book is None:
                print("No such book")
                continue
            try:
                paid = book.sell(ask_int("Copies: "))
            except ValueError as e:
                print("Sale refused:", e)
                continue
            revenue += paid
            print(f"Paid {paid:.2f}. Another sale? {input('(y/n) ')}")
        elif choice == "4":
            print(f"Revenue this session: {revenue:.2f}")
            print("Out of stock:", [t for t, b in books.items() if b.stock == 0])
        elif choice == "5":
            save_catalogue(books, "catalogue.csv")
            print("Saved. Goodbye!")
            break
        else:
            print("Unknown option")

main()
```

## 4. Testing Your Program

A program is finished when it handles **wrong** input as well as right input. Before you submit, test every menu option with normal values, with unusual values and with nonsense. Write small checks with `assert`: if the condition is false, Python stops with an `AssertionError`, so you know exactly which rule is broken.

```python
b = Book("Test Book", 10.0, 20, "Test")

assert b.sell(2) == 23.0
assert b.stock == 18
assert b.sell(10) == 103.5

try:
    b.sell(100)
    print("Problem: an impossible sale was accepted")
except ValueError:
    print("Impossible sale correctly refused")

print("All tests passed")
```

| Test | What to try | Expected result |
| --- | --- | --- |
| Normal use | Sell 2 copies of a book in stock | Correct total; stock goes down by 2 |
| Edge case | Sell exactly 10 copies | The discount is applied |
| Impossible | Sell more copies than in stock | Refused with a message |
| Bad input | Type `abc` when asked for copies | Asked again; no crash |
| Unknown item | Sell a title that does not exist | A clear message |
| Files | Quit, restart, list the catalogue | The changes are still there |

## Common Mistakes

| Symptom | Mistake | Fix |
| --- | --- | --- |
| Crash | Calling `int(input())` without `try` | Wrap it in a function that asks again, like `ask_int`. |
| Lost work | Forgetting to save before `break` | Save inside the quit option, before leaving the loop. |
| Spaghetti code | All the logic inside one huge `while` loop | Move each action into a function or a method. |
| Wrong results | Testing only with perfect input | Use the test checklist, including nonsense input. |
