## In-Class Exercises

### In-Class Exercise 1: Circle Area

```python
def circle_area(radius):
    return 3.14159 * radius ** 2

for r in [1, 2.5, 10]:
    print(f"Radius {r}: area {circle_area(r):.2f}")
```

### In-Class Exercise 2: Grade Letter

```python
def grade_letter(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    return "F"

for s in [95, 83, 71, 64, 40]:
    print(s, grade_letter(s))
```

### In-Class Exercise 3: Temperature Converter

```python
def to_fahrenheit(celsius):
    return celsius * 9 / 5 + 32

def to_celsius(fahrenheit):
    return (fahrenheit - 32) * 5 / 9

f = to_fahrenheit(37)
print(f"{f:.1f}")
print(f"{to_celsius(f):.1f}")
```

### In-Class Exercise 4: Fix the Function

```python
def average_of_three(a, b, c):
    return (a + b + c) / 3

result = average_of_three(70, 85, 90)
print("Average plus bonus:", result + 5)
```

## Part A: Basic

### Exercise 1: Rectangle Tools

```python
def rectangle_area(width, height):
    return width * height

def rectangle_perimeter(width, height):
    return 2 * (width + height)

print("Area:", rectangle_area(7.5, 4))
print("Perimeter:", rectangle_perimeter(7.5, 4))
```

### Exercise 2: Even or Odd

```python
def is_even(number):
    return number % 2 == 0

for n in range(1, 11):
    print(n, "even" if is_even(n) else "odd")
```

## Part B: Intermediate

### Exercise 3: BMI Calculator

```python
def bmi(weight_kg, height_m):
    return weight_kg / height_m ** 2

def bmi_category(value):
    if value < 18.5:
        return "Underweight"
    elif value < 25:
        return "Normal"
    elif value < 30:
        return "Overweight"
    return "Obese"

for w, h in [(70, 1.75), (95, 1.70)]:
    value = bmi(w, h)
    print(f"{value:.1f} {bmi_category(value)}")
```

### Exercise 4: Count Vowels

```python
def count_vowels(text):
    count = 0
    for ch in text.lower():
        if ch in "aeiou":
            count += 1
    return count

print(count_vowels("Data Science with Python"))
```

## Part C: Challenge

### Exercise 5: Prime Numbers Again

`d * d <= n` is enough: if a number has a divisor larger than its square root, it also has one smaller than it.

```python
def is_prime(n):
    if n < 2:
        return False
    d = 2
    while d * d <= n:
        if n % d == 0:
            return False
        d += 1
    return True

def primes_between(start, end):
    count = 0
    for n in range(start, end + 1):
        if is_prime(n):
            print(n, end=" ")
            count += 1
    print()
    return count

print("Count:", primes_between(1, 100))
print("Count:", primes_between(100, 200))
```

## Part D: Cumulative Project: Campus Bookshop

### Exercise 6: Bookshop Functions

```python
def line_total(price, quantity):
    return price * quantity

def discount_rate(quantity, member):
    rate = 0.10 if quantity >= 10 else 0
    if member:
        rate += 0.05
    return rate

def order_total(price, quantity, member):
    after = line_total(price, quantity) * (1 - discount_rate(quantity, member))
    return round(after * 1.15, 2)

def print_receipt(title, price, quantity, member):
    sub = line_total(price, quantity)
    rate = discount_rate(quantity, member)
    print("=" * 32)
    print(f"{title} x {quantity}")
    print(f"Subtotal:  {sub:>10.2f}")
    print(f"Discount:  {rate * 100:>9.0f}%")
    print(f"Total:     {order_total(price, quantity, member):>10.2f}")
    print("=" * 32)

print_receipt("Python Basics", 45.0, 3, False)
print_receipt("Python Basics", 45.0, 12, True)
```
