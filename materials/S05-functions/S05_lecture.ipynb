## 1. Defining and Calling Functions

A function definition starts with `def`, then the function name, brackets and a colon. The indented block below is the **body**. Defining a function does not run it; the body runs only when you **call** the function by writing its name with brackets.

```python
def function_name():
    body
```

````python
def show_welcome():
    print("Welcome to Data Science with Python")
    print("Session 05: Functions")

show_welcome()
show_welcome()
````

Function names follow the same rules as variable names: lowercase words joined by underscores. Choose a verb that says what the function does, such as `show_welcome`, `calculate_total` or `is_leap_year`.

## 2. Parameters and Arguments

A function becomes useful when it can work on different values. A **parameter** is a variable listed in the brackets of the definition. The value you give when you call the function is an **argument**.

````python
def greet(name):
    print(f"Hello, {name}! Welcome to the course.")

greet("Lina")
greet("Omar")
````

````python
def show_order(title, quantity, price):
    total = quantity * price
    print(f"{quantity} x {title} = {total:.2f}")

show_order("Python Basics", 2, 45.0)
show_order("Statistics Made Simple", 1, 38.5)
````

## 3. Returning Values

The functions so far print their result. Often we want the result itself, so that we can store it, compare it or use it in another calculation. The `return` statement sends a value back to the place where the function was called, and ends the function.

````python
def add_tax(price, rate):
    return price + price * rate

total = add_tax(100, 0.15)
print(total)
print(add_tax(40, 0.15) + add_tax(60, 0.15))
````

````python
def is_leap_year(year):
    return (year % 4 == 0 and year % 100 != 0) or year % 400 == 0

for year in [1900, 2000, 2023, 2024]:
    print(year, is_leap_year(year))
````

````python
def area_wrong(width, height):
    print(width * height)

result = area_wrong(3, 4)
print(result + 1)
````

## 4. Default and Keyword Arguments

A parameter can have a **default value**. If the caller does not give that argument, the default is used. Parameters with defaults must come after the ones without.

When you call a function you can also name the arguments. These **keyword arguments** can be given in any order and make long calls easier to read.

````python
def price_after_discount(price, discount=0.10):
    return price * (1 - discount)

print(price_after_discount(200))
print(price_after_discount(200, 0.25))
print(price_after_discount(discount=0.5, price=80))
````

You have already used keyword arguments: `print("a", "b", sep="-")` uses the keyword argument `sep`, and `round(3.14159, 2)` passes the number of digits by position. Library functions in pandas and scikit-learn have many keyword arguments with sensible defaults.

## 5. Scope: Where Variables Live

A variable created inside a function is **local**: it exists only while the function runs. Code outside the function cannot see it. Variables created outside any function are **global** and can be read inside functions, but changing them from inside a function is a common source of bugs. Pass values in as arguments and return results instead.

````python
def calculate_average(scores_total, count):
    average = scores_total / count
    return average

calculate_average(255, 3)
print(average)
````

## 6. Breaking a Problem into Functions

Good programs are built from small functions that each do one job. The main part of the program then reads almost like English. Each function can also be tested on its own.

````python
def subtotal(price, quantity):
    return price * quantity

def discount_rate(quantity, member):
    rate = 0.10 if quantity >= 10 else 0
    if member:
        rate += 0.05
    return rate

def order_total(price, quantity, member, tax=0.15):
    base = subtotal(price, quantity)
    after_discount = base * (1 - discount_rate(quantity, member))
    return round(after_discount * (1 + tax), 2)

print(order_total(45.0, 3, False))
print(order_total(45.0, 12, True))
````

## Common Mistakes

| Symptom | Mistake | Fix |
| --- | --- | --- |
| Nothing happens | Defining a function but never calling it | Write the name with brackets, such as `show_welcome()`, to run it. |
| TypeError | Doing maths with a function that only prints | Use `return` to send the value back instead of printing it. |
| NameError | Using a local variable outside its function | Store the returned value in a variable outside the function. |
| Wrong result | Arguments given in the wrong order | Check the parameter order, or use keyword arguments. |
