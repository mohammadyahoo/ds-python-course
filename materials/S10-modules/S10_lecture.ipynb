A **module** is a file of Python code, such as functions and constants, that you can **import** into your program. A **library** or package is a collection of modules. NumPy, pandas, Matplotlib and scikit-learn, the tools of the rest of this course, are all libraries that you import.

````python
def median(values):
    ordered = sorted(values)
    return ordered[len(ordered) // 2]

print(median([7, 1, 9, 3, 5]))
````

## 1. Importing Modules

`import math` loads the whole module; you then write `math.` before each name. `from math import sqrt` brings in one name so you can use it directly. `import statistics as st` gives the module a shorter **alias**. Data scientists use standard aliases, such as `import numpy as np` and `import pandas as pd`, which you will meet in Stage 2.

Put all imports at the top of the notebook, in their own cell.

````python
import math
from math import sqrt, pi
import statistics as st

print(math.floor(7.8), math.ceil(7.2))
print(sqrt(144), round(pi, 4))
print(st.mean([4, 8, 6]))
````

| Statement | How you use it | When to use it |
| --- | --- | --- |
| `import math` | `math.sqrt(16)` | Most of the time: the name shows where a tool comes from |
| `from math import sqrt` | `sqrt(16)` | When you use one or two names many times |
| `import pandas as pd` | `pd.read_csv(...)` | For libraries with a standard short alias |

## 2. Useful Standard Modules

````python
import statistics

scores = [72, 88, 95, 64, 88, 79]

print("Mean:", round(statistics.mean(scores), 2))
print("Median:", statistics.median(scores))
print("Mode:", statistics.mode(scores))
print("Std dev:", round(statistics.stdev(scores), 2))
````

## 3. Randomness and Seeds

The `random` module produces random numbers, picks random items and shuffles lists. Data scientists use randomness to simulate experiments, take samples and split data into training and test sets.

Random numbers on a computer come from a formula that starts at a **seed**. If you set the same seed, you get the same "random" results every time. That makes your work **reproducible**: someone else can run your notebook and get exactly your numbers. You will see this idea again as `random_state=42` in scikit-learn.

````python
import random

random.seed(42)
print(random.randint(1, 6))
print(random.choice(["heads", "tails"]))
print(random.sample(range(1, 50), 6))
print(round(random.random(), 3))

cards = ["A", "K", "Q", "J"]
random.shuffle(cards)
print(cards)
````

````python
import random

random.seed(7)
rolls = [random.randint(1, 6) for _ in range(6000)]

for face in range(1, 7):
    share = rolls.count(face) / len(rolls)
    print(face, f"{share:.3f}")
````

## 4. Working with Dates

Dates are everywhere in data: order dates, sign-up dates, sensor times. The `datetime` module knows how many days each month has, handles leap years and can do arithmetic with dates. `strptime` turns text into a date; `strftime` turns a date into text in any format.

````python
from datetime import date, datetime, timedelta

start = date(2025, 9, 14)
end = date(2026, 1, 20)
print((end - start).days, "days")
print(start + timedelta(weeks=18))
print(start.strftime("%A %d %B %Y"))

d = datetime.strptime("14/03/2025 09:30", "%d/%m/%Y %H:%M")
print(d.year, d.month, d.hour, d.strftime("%a"))
````

## 5. Writing Your Own Module

Any `.py` file is a module. If you save your bookshop functions in `bookshop_tools.py`, every notebook in the same folder can import them. In Colab, the cell below writes the file into the notebook's folder; on your own computer you would create the file in an editor.

````python
code = """
TAX_RATE = 0.15

def with_tax(price):
    return round(price * (1 + TAX_RATE), 2)

def discount_rate(quantity, member=False):
    rate = 0.10 if quantity >= 10 else 0
    return round(rate + (0.05 if member else 0), 2)
"""

with open("bookshop_tools.py", "w") as f:
    f.write(code)
````

````python
import bookshop_tools as bt

print(bt.TAX_RATE)
print(bt.with_tax(45.0))
print(bt.discount_rate(12, member=True))
````

## Common Mistakes

| Symptom | Mistake | Fix |
| --- | --- | --- |
| NameError | Using `sqrt` after `import math` | Write `math.sqrt`, or use `from math import sqrt`. |
| ModuleNotFoundError | A misspelt module, or one that is not installed | Check the spelling; install it with `!pip install`. |
| Shadowing | Naming your own file `random.py` or `math.py` | Your file hides the real module; choose another name. |
| Changing results | Random numbers without a seed | Call `random.seed(42)` at the start for repeatable results. |
