### Exercise 3: Age Calculator

Comparing tuples such as `(5, 19) < (5, 20)` compares the month first, then the day.

```python
from datetime import date

def age_on(birth_date, on_date):
    age = on_date.year - birth_date.year
    if (on_date.month, on_date.day) < (birth_date.month, birth_date.day):
        age -= 1
    return age

born = date(2004, 5, 20)
print(age_on(born, date(2025, 5, 19)))
print(age_on(born, date(2025, 5, 20)))
```

### Exercise 4: Coin Flip Streaks

```python
import random
random.seed(11)
flips = [random.choice("HT") for _ in range(1000)]
print("Heads:", flips.count("H") / len(flips))
longest = current = 1
for i in range(1, len(flips)):
    current = current + 1 if flips[i] == flips[i - 1] else 1
    longest = max(longest, current)
print("Longest run:", longest)
```

## Part C: Challenge

### Exercise 5: Monte Carlo Pi

More points give estimates closer to pi. This is the law of large numbers, which you will meet again in Session 27.

```python
import math
import random
random.seed(1)
for n in [1_000, 10_000, 100_000]:
    inside = 0
    for _ in range(n):
        x, y = random.random(), random.random()
        if x ** 2 + y ** 2 <= 1:
            inside += 1
    print(n, 4 * inside / n, round(math.pi, 5))
```

## Part D: Cumulative Project: Campus Bookshop

### Exercise 6: The Bookshop Module

```python
code = """
from datetime import date

TAX_RATE = 0.15

def discount_rate(quantity, member=False):
    rate = 0.10 if quantity >= 10 else 0
    return rate + (0.05 if member else 0)

def order_total(price, quantity, member=False):
    after = price * quantity * (1 - discount_rate(quantity, member))
    return round(after * (1 + TAX_RATE), 2)

def receipt_date(order_date):
    return order_date.strftime("%A %d %B %Y")
"""
with open("shop.py", "w") as f:
    f.write(code)

from datetime import date
import shop

print(shop.order_total(45.0, 3))
print(shop.order_total(45.0, 12, member=True))
print("Campus Bookshop,", shop.receipt_date(date(2025, 3, 14)))
```
