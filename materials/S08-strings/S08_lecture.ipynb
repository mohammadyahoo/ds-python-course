## 2. Cleaning Text

String methods never change the original string, because strings are immutable. They return a **new** string, so you store the result: `name = name.strip()`. Methods can be **chained**: each one works on the result of the one before.

```python
raw = "   lina HADDAD  "

print(repr(raw.strip()))
print(raw.strip().lower())
print(raw.strip().upper())
print(raw.strip().title())
print("2,500,000".replace(",", ""))
```

```python
cities = ["Damascus", "damascus ", "DAMASCUS", " Aleppo", "aleppo"]

clean = [c.strip().title() for c in cities]
print(clean)
print(set(clean))
```

## 3. Searching Text

```python
email = "lina.haddad@university.edu"

print("@" in email)
print(email.endswith(".edu"))
print(email.startswith("omar"))
print(email.find("@"))
print(email.count("a"))
print(email[email.find("@") + 1:])
```

| Method | True when the string | Example that is True |
| --- | --- | --- |
| `isdigit()` | has only digits | `"2025".isdigit()` |
| `isalpha()` | has only letters | `"Lina".isalpha()` |
| `isupper()` | has no lowercase letters | `"SQL".isupper()` |
| `startswith(x)` | begins with `x` | `"DS-01".startswith("DS")` |
| `endswith(x)` | finishes with `x` | `"data.csv".endswith(".csv")` |

## 4. Splitting and Joining

`split()` breaks a string into a list. With no argument it splits on any spaces; with an argument such as `","` it splits on that character. `join()` does the opposite: the string before the dot is the glue placed between the items.

```python
sentence = "Python makes data science fun"
words = sentence.split()
print(words)
print(len(words), "words")

row = "Lina,Damascus,3.6"
name, city, gpa = row.split(",")
print(name, city, float(gpa) + 0.1)
```

```python
words = ["Python", "makes", "data", "science", "fun"]
print(" ".join(words))
print(" | ".join(words))
print("-".join(["2025", "03", "14"]))
```

## 5. Formatting Output

You met `:.2f` in Session 02. A format code can also set a **width** and an **alignment**, so values line up in columns: `<` aligns left, `>` aligns right and `^` centres. A comma adds thousands separators, and `%` shows a fraction as a percentage.

```python
items = [("Latte", 3.5, 128),
         ("Croissant", 2.8, 64),
         ("Orange Juice", 3.8, 9)]

print(f"{'Item':<14}{'Price':>7}{'Sold':>6}")
for name, price, sold in items:
    print(f"{name:<14}{price:>7.2f}{sold:>6}")

print(f"{1234567.891:,.2f}")
print(f"{0.1834:.1%}")
```

| Code | Example | Result |
| --- | --- | --- |
| `:.2f` | `f"{3.14159:.2f}"` | `3.14` |
| `:>8` | `f"{42:>8}"` | `      42` |
| `:<8` | `f"{'ab':<8}|"` | `ab      |` |
| `:^9` | `f"{'mid':^9}"` | `   mid   ` |
| `:,` | `f"{2500000:,}"` | `2,500,000` |
| `:.0%` | `f"{0.75:.0%}"` | `75%` |

## 6. From Messy Text to Clean Data

```python
records = ["  Lina Haddad ; DAMASCUS ; 3.62 ",
           "omar saleh;aleppo;3.1",
           "Sara Nassar ; Homs ;3.95"]

students = []
for line in records:
    name, city, gpa = line.split(";")
    students.append({"name": name.strip().title(),
                     "city": city.strip().title(),
                     "gpa": float(gpa)})

for s in students:
    print(s)
```

## Common Mistakes

| Symptom | Mistake | Fix |
| --- | --- | --- |
| No change | Calling `name.strip()` without storing the result | Strings never change; write `name = name.strip()`. |
| TypeError | Assigning to a character, `word[0] = "X"` | Build a new string, such as `"X" + word[1:]`. |
| ValueError | Unpacking a split into the wrong number of names | Check the separator and the number of fields per line. |
| Wrong count | Comparing text without cleaning it | Compare `a.strip().lower()` with `b.strip().lower()`. |
