## In-Class Exercises

### In-Class Exercise 1: Initials

```python
def initials(full_name):
    return "".join(part[0].upper() + "." for part in full_name.split())

print(initials("lina haddad"))
print(initials("Omar Abdul Karim"))
```

### In-Class Exercise 2: Palindrome

```python
def is_palindrome(text):
    clean = text.replace(" ", "").lower()
    return clean == clean[::-1]

print(is_palindrome("Never odd or even"))
print(is_palindrome("Python"))
```

### In-Class Exercise 3: Email Checker

```python
def is_valid_email(email):
    if email.count("@") != 1:
        return False
    user, domain = email.split("@")
    return len(user) > 0 and "." in domain

for e in ["lina@uni.edu", "omar.uni.edu", "@uni.edu", "sara@uni"]:
    print(e, is_valid_email(e))
```

### In-Class Exercise 4: Price Table

```python
menu = {"Espresso": 2.5, "Latte": 3.5, "Chicken Sandwich": 6.5}
for item, price in menu.items():
    print(f"{item:<18}{price:>6.2f}")
```

### Exercise 1: Username Generator

```python
def make_username(first, last, year):
    return (first[0] + last + str(year)[-2:]).lower()

print(make_username("Lina", "Haddad", 2004))
print(make_username("Omar", "Saleh", 1999))
```

### Exercise 3: Phone Number Cleaner

```python
def clean_phone(text):
    digits = "".join(ch for ch in text if ch.isdigit())
    return digits[-9:]

for p in ["0933-111-222", "0933 111 222", "(0933)111222", "+963933111222"]:
    print(p, "->", clean_phone(p))
```

### Exercise 4: Caesar Cipher

```python
def encrypt(text, shift):
    result = ""
    for ch in text:
        if ch.isalpha():
            result += chr((ord(ch) - 97 + shift) % 26 + 97)
        else:
            result += ch
    return result

secret = encrypt("data science", 3)
print(secret)
print(encrypt(secret, -3))
```

### Exercise 5: Parse a Log File

```python
log = [
    "2025-03-14 08:15:02 GET /home 200 0.12",
    "2025-03-14 08:15:09 GET /courses 200 0.30",
    "2025-03-14 08:16:44 POST /login 401 0.08",
    "2025-03-14 08:17:01 GET /courses 200 0.25",
    "2025-03-14 08:18:30 GET /missing 404 0.05",
]
pages = {}
errors = 0
times = []
for line in log:
    date, time, method, page, status, seconds = line.split()
    pages[page] = pages.get(page, 0) + 1
    if int(status) >= 400:
        errors += 1
    times.append(float(seconds))
print(pages)
print("Errors:", errors)
print(f"Average time: {sum(times) / len(times):.3f}")
```

## Part D: Cumulative Project: Campus Bookshop

### Exercise 6: Bookshop Search

```python
titles = ["Python Basics", "Data Science Handbook", "Statistics Made Simple",
          "SQL in a Week", "Machine Learning Guide", "Python for Data Analysis"]

def search(query):
    q = query.strip().lower()
    return [t for t in titles if q in t.lower()]

def code_for(title):
    letters = "".join(word[0].upper() for word in title.split())
    return f"{letters}-{len(title)}"

print(search("  python "))
print(search("DATA"))
for t in titles:
    print(f"{code_for(t):<8}{t}")
```
