## 1. Comparisons and Boolean Values

In Session 02 you met the `bool` type, which has two values: `True` and `False`. A **comparison** asks a yes or no question about two values, and its answer is always a `bool`.

| Operator | Question | Example | Result |
| --- | --- | --- | --- |
| `==` | Equal to? | `5 == 5` | `True` |
| `!=` | Not equal to? | `5 != 3` | `True` |
| `>` | Greater than? | `5 > 8` | `False` |
| `<` | Less than? | `5 < 8` | `True` |
| `>=` | Greater than or equal to? | `5 >= 5` | `True` |
| `<=` | Less than or equal to? | `9 <= 5` | `False` |

Notice the double equals sign. A single `=` stores a value in a variable. A double `==` asks whether two values are equal.

````python
temperature = 31

print(temperature > 30)
print(temperature == 25)
print(temperature != 25)
````

Strings can be compared too. Two strings are equal only if every character matches, including capital letters.

````python
city = "Amman"

print(city == "Amman")
print(city == "amman")
````

## 2. The if Statement

An `if` statement runs a block of code only when its condition is `True`.

```python
if condition:
    code that runs when the condition is True
```

Three details matter:

1. The line ends with a colon `:`.
2. The lines inside the block are **indented** by four spaces. Colab adds them for you when you press Enter after the colon.
3. The block ends where the indentation ends. The next unindented line always runs.

````python
temperature = 34

if temperature > 30:
    print("It is hot today.")
    print("Drink plenty of water.")

print("Weather check complete.")
````

Change `temperature` to 22 and run the cell again. The two indented lines are skipped, but the last line still runs.

## 3. Adding else

`else` gives the program something to do when the condition is `False`. Exactly one of the two blocks runs.

````python
balance = 120
price = 150

if balance >= price:
    print("Payment accepted.")
else:
    print("Not enough balance.")
````

## 4. More Than Two Cases with elif

When there are more than two possible outcomes, add `elif` (short for "else if") branches. Python checks the conditions from top to bottom and runs **only the first block whose condition is `True`**. The `else` at the end catches every case that no condition matched.

````python
mark = 78

if mark >= 90:
    grade = "A"
elif mark >= 80:
    grade = "B"
elif mark >= 70:
    grade = "C"
elif mark >= 60:
    grade = "D"
else:
    grade = "F"

print(f"A mark of {mark} is grade {grade}.")
````

Because Python stops at the first match, **the order of the branches matters**. The version below checks the lowest threshold first. Every mark of 60 or more matches the first branch, so no student can ever get an A, B or C.

````python
mark = 95

if mark >= 60:
    grade = "D"
elif mark >= 70:
    grade = "C"
elif mark >= 90:
    grade = "A"
else:
    grade = "F"

print(f"A mark of {mark} is grade {grade}.")
````

When conditions overlap like this, check the most specific or the highest threshold first.

## 5. Combining Conditions

Three **logical operators** combine conditions into one:

| Operator | Result is True when | Example |
| --- | --- | --- |
| `and` | Both conditions are True | `age >= 18 and has_ticket` |
| `or` | At least one condition is True | `is_member or total > 200` |
| `not` | The condition is False | `not is_closed` |

````python
age = 20
has_ticket = True

if age >= 18 and has_ticket:
    print("Entry allowed.")
else:
    print("Entry refused.")
````

````python
is_member = False
total = 250

if is_member or total > 200:
    print("Free delivery.")
else:
    print("Delivery costs 15.")
````

Python also lets you write a range check the way you would in mathematics: `18 <= age <= 30` means `age >= 18 and age <= 30`.

````python
age = 25

if 18 <= age <= 30:
    print("Eligible for the young professionals discount.")
````

## 6. Nested Conditions

An `if` statement can sit inside another one. This is called **nesting**, and it is useful when the second question only makes sense after the first is answered.

````python
has_account = True
password_correct = False

if has_account:
    if password_correct:
        print("Welcome back.")
    else:
        print("Wrong password.")
else:
    print("Please create an account.")
````

If you only need one combined answer, `and` is usually clearer than nesting. Use nesting when each level needs its own message, as above.

## 7. Checking Text with in

The keyword `in` checks whether one string appears inside another. It is useful for simple text checks.

````python
email = "lina.haddad@example.com"

if "@" in email:
    print("The email address contains an @ sign.")
else:
    print("The email address is not valid.")
````

## 9. Worked Example: Cinema Ticket Prices

A cinema sets its prices with these rules:

1. Children under 12 pay 20, adults pay 35, and people aged 65 or over pay 25.
2. On Tuesdays every ticket is 30 percent cheaper.
3. Students aged 18 to 25 get a further 5 off, but only on days other than Tuesday.

**Step 1.** Store the customer's details.

````python
age = 21
day = "Friday"
is_student = True
````

**Step 2.** Find the base price from the age. The ranges do not overlap, so an `elif` chain is the right tool.

````python
if age < 12:
    price = 20
elif age >= 65:
    price = 25
else:
    price = 35

print(f"Base price: {price}")
````

**Step 3.** Apply the discounts. The Tuesday discount and the student discount cannot both apply, so they belong in one `if` and `elif` pair.

````python
if day == "Tuesday":
    price = price * 0.7
elif is_student and 18 <= age <= 25:
    price = price - 5

print(f"Final price: {price:.2f}")
````

Try different values in Step 1, such as `age = 70` with `day = "Tuesday"`, and check that the final price follows the rules.
