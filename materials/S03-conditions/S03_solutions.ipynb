### In-Class Exercise 4: Leap Years

The parentheses make the rule easy to read: divisible by 4 but not by 100, or divisible by 400.

```python
year = 2024
is_leap = (year % 4 == 0 and year % 100 != 0) or year % 400 == 0
print(year, is_leap)

year = 1900
is_leap = (year % 4 == 0 and year % 100 != 0) or year % 400 == 0
print(year, is_leap)

year = 2000
is_leap = (year % 4 == 0 and year % 100 != 0) or year % 400 == 0
print(year, is_leap)

year = 2026
is_leap = (year % 4 == 0 and year % 100 != 0) or year % 400 == 0
print(year, is_leap)
```

### Exercise 3: Shipping Cost

```python
weight = 6
destination = "international"

if weight <= 2:
    local_cost = 10
    international_cost = 40
elif weight <= 10:
    local_cost = 25
    international_cost = 90
else:
    local_cost = 50
    international_cost = 180

if destination == "local":
    cost = local_cost
else:
    cost = international_cost

print(f"Shipping cost for {weight} kg ({destination}): {cost}")
```

### Exercise 4: Electricity Bill

```python
usage = 250

if usage <= 160:
    bill = usage * 0.05
elif usage <= 300:
    bill = 160 * 0.05 + (usage - 160) * 0.10
else:
    bill = 160 * 0.05 + 140 * 0.10 + (usage - 300) * 0.15

print(f"Bill for {usage} kWh: {bill:.2f}")
```

### Exercise 5: Triangle Classifier

```python
a = 3
b = 4
c = 5

if a + b <= c or a + c <= b or b + c <= a:
    print("Not a triangle")
else:
    if a == b and b == c:
        print("Equilateral")
    elif a == b or b == c or a == c:
        print("Isosceles")
    else:
        print("Scalene")

    if c >= a and c >= b and a ** 2 + b ** 2 == c ** 2:
        print("It is a right triangle")
```

## Part D: Cumulative Project: Campus Bookshop

### Exercise 6: Discounts and Stock

Change the values in the first block to test the three orders. The third order (25 copies with 20 in stock) prints the stock message only.

```python
title = "Python for Beginners"
price = 45.0
quantity = 12
stock = 20
is_member = True
tax_rate = 0.15

if quantity > stock:
    print(f"Sorry, only {stock} copies available")
else:
    subtotal = price * quantity

    discount_rate = 0
    if quantity >= 10:
        discount_rate = discount_rate + 0.10
    if is_member:
        discount_rate = discount_rate + 0.05

    discount = subtotal * discount_rate
    tax = (subtotal - discount) * tax_rate
    total = subtotal - discount + tax

    print("Campus Bookshop")
    print(f"Book: {title}")
    print(f"Quantity: {quantity}")
    print("-" * 30)
    print(f"Subtotal: {subtotal:.2f}")
    print(f"Discount: {discount:.2f}")
    print(f"Tax: {tax:.2f}")
    print(f"Total: {total:.2f}")
    print("-" * 30)
```

The two discount rules use two separate `if` statements, not `if` and `elif`, because a large order from a member gets **both** discounts.
